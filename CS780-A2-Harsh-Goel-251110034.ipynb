## **Random-Maze Environment**

**In this assignment we will be exploring a variant of the Random Maze Environment (RME) that we have been looking in the lectures.**

The environment is represented as a grid world in Figure 1a. Random maze environment
is a highly stochastic environment with 11 states: two terminal states (a goal state (G) and a hole state (H))
and 9 non-terminal states and a wall in between the environment. The wall behaves similar to the wall on the
periphery of the environment, basically if an agent bumps against the wall, it bounces back. The boundary of
the environment behaves similarly, if an agent hits the boundary it bounces back. The agent receives a reward
of +1 when it lands in the goal state (3) and it receives a reward of -1 when it lands in the hole state (7).
For rest of the transitions there is a reward of -0.04. Essentially the agent has the living cost of -0.04.

image.png

The transitions are stochastic as shown in Figure 1b. In this environment, four actions are possible: left, top, right,
and bottom. For every intended action, there is 80% chance of going in the intended direction and remaining
20% chances of going in either of the orthogonal directions. The 20% chance gets equally distributed between
each of the orthogonal direction. The agent starts from state 8 (S). Assume γ = 0.99 for the problems below.

# Create the Maze Environment

```python
class SlipperyMaze(gym.Env):
    """
    A stochastic grid-based maze environment where actions may result
    in unintended movements due to slippery transitions.
    """

    def __init__(self, config=None):
        """
        Initializes the Slippery Maze Environment.

        Args:
            config (dict or None):
                A configuration dictionary containing environment parameters.
                It includes maze dimensions, number of states and actions,
                start, goal, wall, and hole states, reward values, and
                random seed information.

        Returns:
            None
        """
        super().__init__()
        self.config = {}
        if config is not None:
          for key, value in config.items():
            self.config[key] = value

        self.curr_state = self.config['environment']['start_state']
        self.wall_state = self.config['environment']['wall_state']
        self.total_states = self.config['environment']['total_states']
        self.total_actions = self.config['environment']['total_actions']

        self.index_to_states = list(range(self.total_states + 1))
        self.index_to_states.remove(self.wall_state)
        self.states_to_index = {}

        for idx in range(len(self.index_to_states)):
            state = self.index_to_states[idx]
            self.states_to_index[state] = idx

        self.observation_space = gym.spaces.Discrete(self.total_states)
        self.action_space = gym.spaces.Discrete(self.total_actions)

        self.intended_action_prob = 0.8
        self.orthogonal_action_prob = 0.1
        self.max_steps = self.config['max_steps']

        self._make_environment(self.config['environment'])


     # created this method for my ease to push the states, can do within _make_environment() but reducing the readability there
    def _push_state_after_checks(self, config, curr_state, action, ni, nj, prob):
      (x, y) = config['maze_shape']
      next_state = ni * y + nj

      if ni < 0 or ni >= x or nj < 0 or nj >= y or next_state == self.wall_state:
        rwd = config['living_reward']
        self.P[curr_state][action].append((prob, curr_state, rwd, False))

      else:
        rwd = self._get_reward(next_state)
        is_terminal = self._is_terminated(next_state)
        self.P[curr_state][action].append((prob, next_state, rwd, is_terminal))


    def _make_environment(self, config):
        """
        Creates the transition probability model for the maze.

        Args:
            config (dict):
                Dictionary defining the maze layout, total states,
                total actions, maze shape, wall locations, and reward
                specifications.

        Returns:
            dict:
                A transition dictionary P such that:
                P[state][action] is a list of tuples
                (probability, next_state, reward, terminated),
                describing the stochastic outcome of taking an action
                in a given state.
        """
        (x, y) = config['maze_shape']

        # 0 -> right | 1 -> up | 2 -> left | 3 -> down
        directions = {0: (0,1), 1: (-1,0), 2: (0,-1), 3: (1,0)}
        orthogonal_directions = {0: [1,3], 1: [0,2], 2: [1,3], 3: [0,2]}

        self.P = defaultdict(lambda : defaultdict(list))
        for i in range(x):
          for j in range(y):
            curr_state = i * y + j
            if(curr_state == self.wall_state):
              continue

            # for intended action
            for action in range(config['total_actions']):
              # checking whether curr_state is terminal state
              if curr_state == config['goal_state'] or curr_state == config['hole_state']:
                self.P[curr_state][action].append((1.0, curr_state, 0.0, True))
                continue

              # runs only for non terminal states
              (dx, dy) = directions[action]
              ni = i + dx
              nj = j + dy
              self._push_state_after_checks(config, curr_state, action, ni, nj, self.intended_action_prob)

              ## for orthogonal actions
              for ortho_action in orthogonal_directions[action]:
                (ox, oy) = directions[ortho_action]
                oi = i + ox
                oj = j + oy
                self._push_state_after_checks(config, curr_state, action, oi, oj, self.orthogonal_action_prob)

        return self.P


    def _is_terminated(self, state):
        """
        Checks whether a given state is a terminal state.

        Args:
            state (int):
                The current state index of the agent.

        Returns:
            bool:
                True if the state corresponds to a goal or hole state,
                otherwise False.
        """
        return True if state == self.config['environment']['goal_state'] or state == self.config['environment']['hole_state'] else False

    def _get_reward(self, state):
        """
        Computes the reward associated with a given state.

        Args:
            state (int):
                The state index reached after taking an action.

        Returns:
            float:
                Reward value for the given state. This may represent
                a goal reward, hole penalty, or living reward.
        """
        if state == self.config['environment']['goal_state']:
          reward = self.config['environment']['goal_reward']

        elif state == self.config['environment']['hole_state']:
          reward = self.config['environment']['hole_reward']

        else:
           reward = self.config['environment']['living_reward']

        return reward


    def _get_obs(self):
        """
        Generates the observation returned to the agent.

        Returns:
            dict:
                A dictionary containing:
                - agent: the agent's current state
                - terminals: list of terminal states
        """
        obs = {
            'agent': self.curr_state,
            'terminals': [self.config['environment']['goal_state'], self.config['environment']['hole_state']],
        }

        return obs


    def _get_info(self):
        """
        Provides auxiliary diagnostic information.

        Returns:
            dict:
                A dictionary containing additional information such as
                the Manhattan distance between the agent and the goal state.
        """
        goal_state = self.config['environment']['goal_state']
        curr_state = self.curr_state
        (grid_dim_x, grid_dim_y) = self.config['environment']['maze_shape']

        curr_state_x = curr_state // grid_dim_y
        curr_state_y = curr_state % grid_dim_y

        goal_state_x = goal_state // grid_dim_y
        goal_state_y = goal_state % grid_dim_y

        manhatttan_distance = abs(curr_state_x - goal_state_x) + abs(curr_state_y - goal_state_y)

        info = {
            'manhattan_distance b/w agent & goal state': manhatttan_distance,
        }

        return info

    def reset(self, seed=None, options=None):
        """
        Resets the environment to its initial configuration.

        Args:
            seed (int or None):
                Random seed used to ensure reproducibility.
            options (dict or None):
                Additional reset options (if any).

        Returns:
            observation (dict):
                Initial observation of the environment.
            info (dict):
                Additional information after reset.
        """
        if seed is not None:
          super().reset(seed = seed)
        else:
          super().reset()

        self.curr_state = self.config['environment']['start_state']
        observation = self._get_obs()
        info = self._get_info()
        self.curr_step = 0

        return observation, info


    def step(self, action):
        """
        Executes one time step within the environment.

        Args:
            action (int):
                Action selected by the agent. Represents a movement
                direction in the maze.

        Returns:
            observation (dict):
                Observation after performing the action.
            reward (float):
                Reward received after the action.
            terminated (bool):
                Whether the episode has reached a terminal state.
            truncated (bool):
                Whether the episode was truncated due to time limits.
            info (dict):
                Additional diagnostic information including transition logs.
        """
        curr_state = self.curr_state
        transition = self.P[curr_state][action]

        # for rest of the states, there are three possible states possible first with 0.8, 0.1 , 0.1 probabilities respectively
        prob_of_next_states = np.array([p for p, _, _, _ in transition], dtype=np.float32)
        prob_of_next_states /= prob_of_next_states.sum()

        index = self.np_random.choice(len(transition), p = prob_of_next_states)
        _, next_state, reward, terminated = transition[index]

        self.curr_state = next_state
        observation = self._get_obs()
        info = self._get_info()
        self.curr_step += 1
        truncated = self.curr_step >= self.max_steps

        return observation, reward, terminated, truncated, info


    def render(self):
        """
        Renders the current state of the environment.

        Returns:
            None
        """
        self._render_frame()


    def _render_frame(self):
        """
        Renders a single frame of the environment for visualization.

        Returns:
            None
        """
        (grid_dim_x, grid_dim_y) = self.config['environment']['maze_shape']

        goal = self.config['environment']['goal_state']
        hole = self.config['environment']['hole_state']
        (gx, gy) = goal // grid_dim_y, goal % grid_dim_y
        (hx, hy) = hole // grid_dim_y, hole % grid_dim_y

        curr_state = self.curr_state
        (cx, cy) = curr_state // grid_dim_y, curr_state % grid_dim_y

        wall_state = self.config['environment']['wall_state']
        (wx, wy) = wall_state // grid_dim_y, wall_state % grid_dim_y

        agent_state_grid = np.empty((grid_dim_x, grid_dim_y), dtype=str)
        agent_state_grid.fill('⬜')
        agent_state_grid[gx][gy] = '🏆'
        agent_state_grid[hx][hy] = '🕳️'
        agent_state_grid[cx][cy] = '🤖'
        agent_state_grid[wx][wy] = '🧱'

        for row in agent_state_grid:
          print(" ".join(row))
        print()


    def close(self):
        """
        Cleans up environment resources.

        Returns:
            None
        """
        del self.P
        del self.action_space
        del self.observation_space
        del self.config
        del self.curr_state
        del self.intended_action_prob
        del self.orthogonal_action_prob
```

### Finding true state-values \& action-state-values.

```python
def true_values(env):
  env = env.unwrapped
  config = env.config
  gamma = config['discount_factor']
  total_states = config['environment']['total_states']
  total_actions = config['environment']['total_actions']

  Vopt = np.zeros(total_states, dtype=float)
  Qopt = np.zeros(shape=(total_states, total_actions), dtype=float)

  while True:
    Vold = Vopt.copy()
    Qold = Qopt.copy()
    Qopt.fill(0)

    for stateidx in range(total_states):
      for action in range(total_actions):
        state = env.index_to_states[stateidx]
        transitions = env.P[state][action]
        for (prob, next_state, reward, terminated) in transitions:
          if terminated:
            Qopt[stateidx][action] += prob * reward
          else:
            next_state_idx = env.states_to_index[next_state]
            Qopt[stateidx][action] += prob * (reward + gamma * np.max(Qold[next_state_idx]))

      Vopt[stateidx] = np.max(Qopt[stateidx])

    if np.max(np.abs(Vopt - Vold)) < 1e-8:
      break

  pi_opt = np.argmax(Qopt, axis=1)
  return Vopt, Qopt, pi_opt
```

### Generate Trajectory based on Policy $\pi$ and return experience tuples $(s, a, r, s^{\prime})$

```python
def generate_episode_trajectory(environment, Q, epsilon=None, config=None):
    """
    Generates a single episode trajectory by interacting with the environment
    using an epsilon-greedy policy derived from a given Q-function.

    Args:
        environment (gym.Env):
            The environment in which the agent operates. It must support
            reset() and step() methods and define a discrete action space.

        Q (numpy.ndarray):
            The action-value function (Q-table) of shape
            (number_of_states, number_of_actions). It is used to select
            greedy actions during exploitation.

        epsilon (float or None):
            Probability of selecting a random action instead of the greedy
            action. If None, the value is taken from the configuration.

        config (dict):
            Configuration dictionary containing episode-level parameters.
            This includes:
            - random seed
            - exploration rate
            - maximum number of steps per episode
            - environment-specific parameters

    Returns:
        list:
            A list of tuples representing the episode trajectory.
            Each tuple has the form:
            (current_state, action, reward, next_state)

            If the episode exceeds the maximum allowed number of steps,
            an empty list is returned.
    """
    epsilon = epsilon if epsilon is not None else config['epsilon']
    gamma = config['discount_factor']
    trajectory = []

    observation, info = environment.reset()
    curr_state = observation['agent']

    base_env = environment.unwrapped
    terminated = False
    truncated = False
    while not terminated and not truncated:
      if environment.np_random.random() > epsilon:
        action = np.argmax(Q[base_env.states_to_index[curr_state]])
      else:
        action = environment.action_space.sample()

      observation, reward, terminated, truncated, info = environment.step(action)
      next_state = observation['agent']

      trajectory.append((base_env.states_to_index[curr_state], action, reward, base_env.states_to_index[next_state]))
      curr_state = next_state

    return trajectory if terminated else []
```

### Decay $\alpha$

```python
def decay_step_size(initial_value, final_value, episode, max_episode, decay_type, decay_stop=None):
    """
    Computes the step size (learning rate) for a given episode based on
    a specified decay strategy.

    Args:
        initial_value (float):
            Initial step size at the beginning of training.

        final_value (float):
            Final step size to which the learning rate decays.

        episode (int):
            Current episode index for which the step size is computed.

        max_episode (int):
            Total number of episodes over which decay is applied.

        decay_type (str):
            Type of decay strategy used for step size scheduling.
            Supported values include 'exponential' and 'linear'.

        decay_stop (int or None):
            Optional episode index after which the step size remains
            constant at the final value.

    Returns:
        float:
            Step size corresponding to the given episode based on the
            chosen decay strategy.
    """
    eps = 1e-8

    if decay_type == 'linear':
      if decay_stop == None or decay_stop >= episode:
        step_size = initial_value + episode * (final_value - initial_value) / (max_episode - 1 + eps)
      else:
        step_size = final_value

    else:
      if decay_stop == None or decay_stop >= episode:
        decay_factor = np.exp((np.log(final_value + eps) - np.log(initial_value + eps)) / (max_episode - 1 + eps))
        step_size = initial_value * decay_factor**episode
      else:
        step_size = final_value

    return step_size
```

#### Function: generate_episode_step_sizes

```python
def generate_episode_step_sizes(initial_value, final_value, max_episode, decay_type, decay_stop=None):
    """
    Generates step sizes for all episodes using a specified decay strategy.

    Args:
        initial_value (float):
            Initial step size at the start of training.

        final_value (float):
            Final step size at the end of decay.

        max_episode (int):
            Total number of episodes for which step sizes are generated.

        decay_type (str):
            Decay strategy used to compute step sizes
            (e.g., 'exponential' or 'linear').

        decay_stop (int or None):
            Optional episode index after which the step size
            remains fixed.

    Returns:
        list:
            A list of step size values, where each entry corresponds
            to the step size for a particular episode.
    """
    step_sizes = []
    for episode in range(max_episode):
      step_sizes.append(decay_step_size(initial_value, final_value, episode, max_episode, decay_type, decay_stop))

    return step_sizes
```

### Decay $\epsilon$

```python
def decay_epsilon(initial_value, final_value, episode, max_episode, decay_type, decay_stop=None):
    """
    Computes the exploration rate (epsilon) for a given episode based on
    a specified decay strategy.

    Args:
        initial_value (float):
            Initial value of epsilon at the beginning of training.

        final_value (float):
            Minimum value of epsilon after decay.

        episode (int):
            Current episode index for which epsilon is computed.

        max_episode (int):
            Total number of episodes over which epsilon decay is applied.

        decay_type (str):
            Type of decay strategy used for epsilon scheduling.
            Supported values include 'exponential' and 'linear'.

        decay_stop (int or None):
            Optional episode index after which epsilon remains
            fixed at the final value.

    Returns:
        float:
            Epsilon value corresponding to the given episode based on
            the chosen decay strategy.
    """
    eps = 1e-8

    if decay_type == 'linear':
      if decay_stop == None or decay_stop >= episode:
        epsilon = initial_value + episode * (final_value - initial_value) / (max_episode - 1 + eps)
      else:
        epsilon = final_value

    else:
      if decay_stop == None or decay_stop >= episode:
        decay_factor = np.exp((np.log(final_value + eps) - np.log(initial_value + eps)) / (max_episode - 1 + eps))
        epsilon = initial_value * decay_factor**episode
      else:
        epsilon = final_value

    return epsilon
```

#### Function: generate_episode_epsilons

```python
def generate_episode_epsilons(initial_value, final_value, max_episode, decay_type, decay_stop=None):
    """
    Generates epsilon values for all episodes using a specified decay strategy.

    Args:
        initial_value (float):
            Initial epsilon value at the start of training.

        final_value (float):
            Final epsilon value after decay.

        max_episode (int):
            Total number of episodes for which epsilon values are generated.

        decay_type (str):
            Decay strategy used for computing epsilon values
            (e.g., 'linear' or 'exponential').

        decay_stop (int or None):
            Optional episode index after which epsilon remains constant.

    Returns:
        list:
            A list of epsilon values, where each element corresponds
            to the exploration rate for a particular episode.
    """
    epsilons = []
    for episode in range(max_episode):
      epsilons.append(decay_epsilon(initial_value, final_value, episode, max_episode, decay_type, decay_stop))

    return epsilons
```

### Func: get_policy_sucess(env, policy, config)

```python
def get_policy_success(environment, policy, config=None):
    """
    Evaluates whether a given policy successfully reaches the goal state
    within a specified number of steps and computes the discounted return.

    Args:
        environment (gym.Env):
            The environment in which the policy is evaluated. It must
            support reset() and step() operations.

        policy (list or numpy.ndarray):
            A deterministic policy mapping each state to an action.
            The index represents the state and the value represents
            the action selected in that state.

        config (dict):
            Configuration dictionary containing evaluation parameters.
            This includes:
            - random seed
            - maximum number of steps
            - discount factor
            - environment-specific goal state information

    Returns:
        success (bool):
            True if the policy reaches the goal state within the allowed
            number of steps, otherwise False.

        return_reward (float):
            Discounted cumulative reward obtained while following
            the policy during the episode.
    """
    # defined some useful variables
    env = environment
    base_env = env.unwrapped
    observation, info = env.reset(seed = config['seed'])
    s = observation['agent']
    sidx = base_env.states_to_index[s]

    gamma = config['discount_factor']
    max_steps = config['max_steps']
    goal_state = config['environment']['goal_state']
    terminated = False
    truncated = False
    discounted_cumm_reward = 0.0
    step = 0

    while not truncated and not terminated:
      action = policy[sidx]
      observation, reward, terminated, truncated, _ = env.step(action)

      s_prime = observation['agent']
      discounted_cumm_reward += ((gamma**step) * reward)

      sidx = base_env.states_to_index[s_prime]
      s = s_prime
      step += 1

    return True if s == goal_state else False, discounted_cumm_reward
```

```python
def run_tests(n_tests=50, method=None, config=None):
    """
    Executes multiple independent experimental trials for a given
    reinforcement learning method and aggregates the results.

    Args:
        n_tests (int):
            Number of independent test runs to be performed. Each test
            uses a different random seed to evaluate robustness and
            stability of the learning method.

        method (callable):
            A reinforcement learning control method to be evaluated.
            This function is expected to take an environment and a
            configuration dictionary as input and return learned
            value functions, policies, and performance metrics.

        config (dict):
            Configuration dictionary containing experiment and
            environment parameters. This includes:
            - number of episodes
            - environment settings
            - learning hyperparameters
            - random seed information

    Returns:
        Qs (numpy.ndarray):
            A 4D array containing Q-value estimates for all tests,
            episodes, states, and actions.

        V_e (numpy.ndarray):
            A 3D array containing state-value estimates for all tests,
            episodes, and states.

        policy_success (numpy.ndarray):
            A 2D boolean array indicating whether the learned policy
            successfully reached the goal state in each episode
            across all tests.

        return_reward (numpy.ndarray):
            A 2D array containing the discounted return obtained
            in each episode across all test runs.
    """
    max_episodes = config['max_episodes']
    total_actions = config['environment']['total_actions']
    total_states = config['environment']['total_states']

    Qs = np.zeros(shape=(n_tests, max_episodes, total_states, total_actions))
    V_e = np.zeros(shape=(n_tests, max_episodes, total_states))
    policy_success = np.zeros(shape=(n_tests, max_episodes))
    return_reward = np.zeros(shape=(n_tests, max_episodes))
    base_seed = config['seed']

    for test_num in range(n_tests):
      # updating seed within config['seed'] to be used by the method (callable)
      seed = base_seed + test_num
      config['seed'] = seed

      env = gym.make(
          'my_slippery_maze',
          config = config,
      )

      env.reset(seed=config['seed'])
      env.action_space.seed(config['seed'])
      env.observation_space.seed(config['seed'])

      _, Ve_e, _, Qe_s, _, policy_success_e, return_reward_e = method(env, config)
      Qs[test_num] = Qe_s
      V_e[test_num] = Ve_e
      policy_success[test_num] = policy_success_e
      return_reward[test_num] = return_reward_e
      env.close()

    return Qs, V_e, policy_success, return_reward
```

# Problem 1: Monte Carlo Control (40 points)
($40+20+20+5+5+5+5=100$ points)

Implement the Monte Carlo Control for the Random Maze Environment (RME) described above. In particular, you need to implement First Visit Monte Carlo Control (FMVCC) for finding the optimal policy for
RME. Use the function definition (given below) as given in Lecture slides.
MonteCarloControl(env, $\gamma$, $\alpha_0$, $\epsilon_0$, maxSteps, noEpisodes, firstVisit = True)


```python
def monte_carlo_control(environment, config=None):
    """
    Implements Monte Carlo Control for a discrete Random Maze Environment (RME).

    This function applies First-Visit Monte Carlo Control (FMVCC) to learn the
    optimal action-value function Q(s, a) and the corresponding optimal policy.
    An epsilon-greedy policy is used for exploration, with optional decay of
    epsilon and step size across episodes.

    Args:
        environment (gym.Env): Environment with discrete observation and action spaces.
        config (dict): Configuration dictionary containing:
            - max_episodes (int): Number of Monte Carlo episodes.
            - discount_factor (float): Discount factor γ.
            - step_size (dict): Step-size (alpha) decay parameters.
            - epsilon (dict): Epsilon-greedy decay parameters.
            - monte_carlo_strategy (str): 'first_visit' or 'every_visit'.

    Returns:
        Q (np.ndarray): Final learned action-value function of shape (S, A).
        V_e (np.ndarray): State-value estimates per episode.
        policy (np.ndarray): Final greedy policy derived from Q.
        Qs (np.ndarray): History of Q-values across episodes.
        mc_target_tensor (np.ndarray): Monte Carlo return targets for each (s, a).
        policy_success (np.ndarray): Boolean success indicator per episode.
        return_reward (np.ndarray): Total return obtained per episode.
    """

    # Reseting the seed for reproducibility
    env = environment
    np.random.seed(config['seed'])
    random.seed(config['seed'])
    env.reset(seed=config['seed'])

    # Extracting required data from config
    max_episodes = config['max_episodes']
    mc_strategy = config['monte_carlo_strategy']
    gamma = config['discount_factor']
    total_states = config['environment']['total_states']
    total_actions = config['environment']['total_actions']
    goal_state = config['environment']['goal_state']

    # Generating the episode-wise step sizes for 'max_episodes' number of episodes
    # See Section: generate_episode_step_sizes (above)
    alphadecay = generate_episode_step_sizes(**config['step_size'], max_episode=max_episodes)

    # Generating the episode-wise epsilons for 'max_episodes' number of episodes
    # See Section: generate_episode_epsilons (above)
    epsilondecay = generate_episode_epsilons(**config['epsilon'], max_episode=max_episodes)

    # Defining variables to be return in the end as mentioned in the function documentation above
    Q = np.zeros(shape=(total_states, total_actions), dtype=np.float32)
    V_e = np.zeros(shape=(max_episodes, total_states), dtype=np.float32)
    policy = np.empty(total_states)
    Qs = np.zeros(shape=(max_episodes, total_states, total_actions), dtype=np.float32)
    mc_target_tensor = np.zeros(shape=(max_episodes, total_states, total_actions), dtype=np.float32)
    policy_success = np.zeros(max_episodes, dtype=np.float32)
    return_reward = np.zeros(max_episodes, dtype=np.float32)

    # MC control logic
    for episode in range(max_episodes):
        # Extracting alpha and epsilon for the current episode fron alphadecay and epsilondecay respectively
        alpha = alphadecay[episode]
        epsilon = epsilondecay[episode]

        # Generating the trajectory for the current episode
        # See Section: Generate Trajectory based on Policy
        trajectory = generate_episode_trajectory(env, Q, epsilon, config)

        tlen = len(trajectory)
        G = [0] * tlen
        G_t = 0

        '''
        Since MC Control uses FVMC so first getting the `estimated_expected_return`(G_t) from state `s` to the end of the trajectory by iterating from backward

        G_t Forward equation:
        G_t = r_{t+1} + gamma * r_{t+2} + gamma^2 * r_{t+3} + ...

        G_t Backward equation:
        G_t = r_{t+1} + gamma * G_{t+1}

        In Forward equation, we have to do heavy computation by visiting all <state,action> ahead of current <state,action> in a trajectory to find `G_t`
        So to save computation, first traversing over the trajectory from backward and computing estimated_expected_return for each <state,action> pair in one pass
        '''
        for i in range(tlen-1, -1, -1):
            s, a, r, ns = trajectory[i]
            G_t = r + gamma * G_t
            G[i] = G_t

        # creating a set of <visited states, corresponding action> pair so that we can skip if any <state,action> pair is occuring multiple times in a trajectory
        visited = set()

        '''
        updating action values for all <state,action> pair traversed during the trajectory but only at once for each pair as in FVMC, we update Q only for first occurence of <state,action> pair.

        Monte Carlo Update Rule (Incremental Form):
        Q(s,a) = Q(s,a) + alpha * [ G_t - Q(s,a) ]

        where:
          G_t = r_{t+1} + gamma * r_{t+2} + gamma^2 * r_{t+3} + ...
        '''
        for i in range(tlen):
            s, a, r, ns = trajectory[i]
            if (s, a) in visited:
                continue

            visited.add((s, a))
            Q[s][a] += alpha * (G[i] - Q[s][a])
            mc_target_tensor[episode][s][a] = G[i]

        V_e[episode] = np.max(Q, axis=1)
        Qs[episode] = Q.copy()

        # Evaluating Performance Metrics in each episode
        policy = np.argmax(Q, axis=1)
        for _ in range(config['rollouts']):
          # Finding policySucess and returnReward for each rollout using `get_policy_sucess(env,policy,config)`
          # See Section: get_policy_success(env, policy, config)
          policySuccess, returnReward = get_policy_success(env, policy, config)

          # adding policySucess of all rollouts within each episode
          policy_success[episode] += policySuccess
          # adding returnReward of all rollouts within each episode
          return_reward[episode] += returnReward

        # Taking average of policy_success and return_reward for number of rollouts used
        policy_success[episode] /= config['rollouts']
        return_reward[episode] /= config['rollouts']

    # Finally returning the mentioned variables above in docstring
    return Q, V_e, policy, Qs, mc_target_tensor, policy_success, return_reward
```

# Problem 2: SARSA (TD Control)
($40+20+20+5+5+5+5=100$ points)

Implement the SARSA algorithm for the Random Maze Environment (RME) described above. Use the function definition as given in Lecture slides.
SARSA(env,$\gamma$, $\alpha_0$, $\epsilon_0$, noEpisodes)(40 Points)



```python
def sarsa_control(environment, config=None):
    """
    Implements the SARSA (State–Action–Reward–State–Action) algorithm for a
    discrete Random Maze Environment (RME).

    This function applies the on-policy Temporal Difference (TD) control method
    SARSA to learn the optimal action-value function Q(s, a). An epsilon-greedy
    policy is used for both action selection and policy evaluation, with optional
    decay of the learning rate (alpha) and exploration rate (epsilon) across
    episodes.

    Args:
        environment (gym.Env): Environment with discrete observation and action spaces.
        config (dict): Configuration dictionary containing:
            - max_episodes (int): Number of training episodes.
            - discount_factor (float): Discount factor γ.
            - step_size (dict): Learning rate (alpha) decay parameters.
            - epsilon (dict): Epsilon-greedy exploration decay parameters.
            - seed (int): Random seed for environment reset.

    Returns:
        Q (np.ndarray): Final learned action-value function of shape (S, A).
        V_s (np.ndarray): State-value estimates derived from Q for each episode.
        policy (np.ndarray): Final greedy policy derived from Q.
        Qs (np.ndarray): History of Q-values across episodes.
        td_target (float): Last computed TD target value.
        policy_success (np.ndarray): Boolean success indicator per episode.
        return_reward (np.ndarray): Total return obtained per episode.
    """

    # Reseting the seed for reproducibility
    env = environment
    np.random.seed(config['seed'])
    random.seed(config['seed'])
    env.reset(seed=config['seed'])

    # Unwrapping the environment for mapping either states to index ot index to states
    # (Since I haven't register the wall state in the observation space so handling the mapping explicitly)
    base_env = env.unwrapped

    # Extracting required data from config
    max_episodes = config['max_episodes']
    gamma = config['discount_factor']
    total_states = config['environment']['total_states']
    total_actions = config['environment']['total_actions']
    goal_state = config['environment']['goal_state']

    # Generating the episode-wise step sizes for 'max_episodes' number of episodes
    # See Section: generate_episode_step_sizes (above)
    alphadecay = generate_episode_step_sizes(**config['step_size'], max_episode=max_episodes)

    # Generating the episode-wise epsilons for 'max_episodes' number of episodes
    # See Section: generate_episode_epsilons (above)
    epsilondecay = generate_episode_epsilons(**config['epsilon'], max_episode=max_episodes)

    # Defining variables to be return in the end as mentioned above in docstring
    Q = np.zeros(shape=(total_states, total_actions), dtype=np.float32)
    V_s = np.zeros(shape=(max_episodes, total_states), dtype=np.float32)
    policy = np.empty(total_states)
    Qs = np.zeros(shape=(max_episodes, total_states, total_actions))
    td_target = 0.0
    policy_success = np.zeros(max_episodes, dtype=np.float32)
    return_reward = np.zeros(max_episodes, dtype=np.float32)

    # SARSA control logic
    for e in range(max_episodes):
      alpha = alphadecay[e]
      epsilon = epsilondecay[e]
      obs, info = env.reset()

      # mapping state to index (obs returns dict of agent storing current state and list of terminals)
      s = base_env.states_to_index[obs['agent']]

      # actionSelect takes index of the state not actual state
      # See section: Action Selection Logic
      a = actionSelect(env, s, Q, epsilon)

      done = False
      truncated = False
      '''
      SARSA uses bootstrapping to update its action-value as mentioned below:
      Q(s,a) = Q(s,a) + alpha * [ r + gamma * Q(s',a') - Q(s,a) ]

      where:
      td_target = r + gamma * Q(s',a') and
      td_error = td_target - Q(s,a)
      '''
      while not done and not truncated:
        obs_prime, r, done, truncated, info = env.step(a)
        s_prime = base_env.states_to_index[obs_prime['agent']]
        a_prime = actionSelect(env, s_prime, Q, epsilon)

        td_target = r
        if not done:
          td_target += gamma * Q[s_prime][a_prime]
        td_error = td_target - Q[s][a]

        Q[s][a] += alpha * td_error
        a = a_prime
        s = s_prime

      Qs[e] = Q.copy()
      V_s[e] = np.max(Q, axis=1)

      # Evaluating Performance Metrics in each episode
      policy = np.argmax(Q, axis=1)
      for _ in range(config['rollouts']):
        # Finding policySucess and returnReward for each rollout using `get_policy_sucess(env,policy,config)`
        # See Section: get_policy_success(env, policy, config)
        policySuccess, returnReward = get_policy_success(env, policy, config)

        # adding policySucess of all rollouts within each episode
        policy_success[e] += policySuccess
        # adding returnReward of all rollouts within each episode
        return_reward[e] += returnReward

      # Taking average of policy_success and return_reward for number of rollouts used
      policy_success[e] /= config['rollouts']
      return_reward[e] /= config['rollouts']

    # Finally returning the mentioned variables above in docstring
    return Q, V_s, policy, Qs, td_target, policy_success, return_reward
```

# Problem 3: Q-Learning
($40+20+20+5+5+5+5=100$ points)

Implement the Q-Learning algorithm for the Random Maze Environment (RME) described above. Use the function definition as given in Lecture slides.
Q-Learning(env, $\gamma$, $\alpha_0$, $\epsilon_0$, noEpisodes) (40 Points)

```python
def q_learning_control(environment, config=None):
    """
    Implements the Q-Learning algorithm for a discrete Random Maze Environment (RME).

    This function applies Q-Learning, an off-policy Temporal Difference (TD) control
    algorithm, to learn the optimal action-value function Q(s, a). Action selection
    is performed using an epsilon-greedy behavior policy, while updates are based on
    the greedy target policy. Learning rate (alpha) and exploration rate (epsilon)
    may decay across episodes.

    Args:
        environment (gym.Env): Environment with discrete observation and action spaces.
        config (dict): Configuration dictionary containing:
            - max_episodes (int): Number of training episodes.
            - discount_factor (float): Discount factor γ.
            - step_size (dict): Learning rate (alpha) decay parameters.
            - epsilon (dict): Epsilon-greedy exploration decay parameters.
            - seed (int): Random seed for environment reset.

    Returns:
        Q (np.ndarray): Final learned action-value function of shape (S, A).
        V_s (np.ndarray): State-value estimates derived from Q for each episode.
        policy (np.ndarray): Final greedy policy derived from Q.
        Qs (np.ndarray): History of Q-values across episodes.
        None: Placeholder (no TD target history stored).
        policy_success (np.ndarray): Boolean success indicator per episode.
        return_reward (np.ndarray): Total return obtained per episode.
    """

    # Resetting the seed for reproducibility
    env = environment
    np.random.seed(config['seed'])
    random.seed(config['seed'])
    env.reset(seed=config['seed'])

    # Unwrapping the environment for mapping either states to index ot index to states
    # (Since I haven't register the wall state in the observation space so handling the mapping explicitly)
    base_env = env.unwrapped

    # Extracting required data from config
    max_episodes = config['max_episodes']
    gamma = config['discount_factor']
    total_states = env.observation_space.n
    total_actions = env.action_space.n
    goal_state = config['environment']['goal_state']

    # Generating the episode-wise step sizes for 'max_episodes' number of episodes
    # See Section: generate_episode_step_sizes (above)
    alphadecay = generate_episode_step_sizes(**config['step_size'], max_episode=max_episodes)

    # Generating the episode-wise epsilons for 'max_episodes' number of episodes
    # See Section: generate_episode_epsilons (above)
    epsilondecay = generate_episode_epsilons(**config['epsilon'], max_episode=max_episodes)

    # defining variables to be return in the end as mentioned above in docstring
    Q = np.zeros(shape=(total_states, total_actions), dtype=np.float32)
    V_s = np.zeros(shape=(max_episodes, total_states), dtype=np.float32)
    policy = np.empty(total_states)
    Qs = np.zeros(shape=(max_episodes, total_states, total_actions))
    td_target = 0.0
    return_reward = np.zeros(max_episodes, dtype=np.float32)
    # policy_success returns the policy_success rate which is evaluated using policy after each episode
    policy_success = np.zeros(max_episodes, dtype=np.float32)


    # Q-Learning Logic
    for e in range(max_episodes):
      alpha = alphadecay[e]
      epsilon = epsilondecay[e]
      obs, info = env.reset()
      # mapping state to index (obs returns dict of agent storing current state and list of terminals)
      s = base_env.states_to_index[obs['agent']]

      done = False
      truncated = False
      '''
      Q-Learning uses off-policy bootstrapping strategy that directly approximates the optimal policy as mentioned below:
      Q(s,a) = Q(s,a) + alpha * [ r + gamma * max_a'(Q(s',a')) - Q(s,a) ]

      where:
      td_target = r + gamma * max_a'(Q(s',a')) and
      td_error = td_target - Q(s,a)
      '''
      while not done and not truncated:
        a = actionSelect(env, s, Q, epsilon)
        obs_prime, r, done, truncated, info = env.step(a)
        s_prime = base_env.states_to_index[obs_prime['agent']]

        td_target = r
        if not done:
          td_target += gamma * np.max(Q[s_prime])
        td_error = td_target - Q[s][a]

        Q[s][a] += alpha * td_error
        s = s_prime

      Qs[e] = Q.copy()
      V_s[e] = np.max(Q, axis=1)

      # Evaluating Performance Metrics in each episode
      policy = np.argmax(Q, axis=1)
      for _ in range(config['rollouts']):
        # Finding policySucess and returnReward for each rollout using `get_policy_sucess(env,policy,config)`
        # See Section: get_policy_success(env, policy, config)
        policySuccess, returnReward = get_policy_success(env, policy, config)

        # adding policySucess of all rollouts within each episode
        policy_success[e] += policySuccess
        # adding returnReward of all rollouts within each episode
        return_reward[e] += returnReward

      # Taking average of policy_success and return_reward for number of rollouts used
      policy_success[e] /= config['rollouts']
      return_reward[e] /= config['rollouts']

    # Finally returning the mentioned variables above in docstring
    return Q, V_s, policy, Qs, None, policy_success, return_reward
```

# **Problem 4: Double Q-Learning**

($40+20+20+5+5+5+5=100$ points)

Implement the Double Q-Learning algorithm for the Random Maze Environment (RME) described above.
Use the function definition as given in Lecture slides.

Double-Q-Learning(env, $\gamma$, $\alpha_0$, $\epsilon_0$, noEpisodes) (40 Points)

```python
def double_q_learning_control(environment, config=None):
    """
    Implements the Double Q-Learning algorithm for a discrete Random Maze
    Environment (RME).

    This function applies Double Q-Learning, an off-policy Temporal Difference
    (TD) control algorithm designed to reduce maximization bias present in
    standard Q-Learning. Two independent action-value functions (Q1 and Q2)
    are learned by alternating updates, while action selection is performed
    using an epsilon-greedy policy derived from their average.

    Args:
        environment (gym.Env): Environment with discrete observation and action spaces.
        config (dict): Configuration dictionary containing:
            - max_episodes (int): Number of training episodes.
            - discount_factor (float): Discount factor γ.
            - step_size (dict): Learning rate (alpha) decay parameters.
            - epsilon (dict): Epsilon-greedy exploration decay parameters.
            - seed (int): Random seed for environment reset.

    Returns:
        Q (np.ndarray): Final combined action-value function (average of Q1 and Q2).
        V_s (np.ndarray): State-value estimates derived from Q for each episode.
        policy (np.ndarray): Final greedy policy derived from Q.
        Qs (np.ndarray): History of combined Q-values across episodes.
        None: Placeholder (no TD target history stored).
        policy_success (np.ndarray): Boolean success indicator per episode.
        return_reward (np.ndarray): Total return obtained per episode.
    """
    # Resetting the seed for reproducibility
    env = environment
    np.random.seed(config['seed'])
    random.seed(config['seed'])
    env.reset(seed=config['seed'])

    # Unwrapping the environment for mapping either states to index ot index to states
    # (Since I haven't register the wall state in the observation space so handling the mapping explicitly)
    base_env = env.unwrapped

   # Extracting required data from config
    max_episodes = config['max_episodes']
    gamma = config['discount_factor']
    total_states = config['environment']['total_states']
    total_actions = config['environment']['total_actions']
    goal_state = config['environment']['goal_state']
    hole_state = config['environment']['hole_state']
    terminal_states = [goal_state, hole_state]

    # Generating the episode-wise step sizes for 'max_episodes' number of episodes
    # See Section: generate_episode_step_sizes (above)
    alphadecay = generate_episode_step_sizes(**config['step_size'], max_episode=max_episodes)

    # Generating the episode-wise epsilons for 'max_episodes' number of episodes
    # See Section: generate_episode_epsilons (above)
    epsilondecay = generate_episode_epsilons(**config['epsilon'], max_episode=max_episodes)

    # defining variables to be return in the end as mentioned above in docstring
    Q1 = np.zeros(shape=(total_states, total_actions), dtype=np.float32)
    Q2 = np.zeros(shape=(total_states, total_actions), dtype=np.float32)
    Q = (Q1 + Q2) / 2
    V_s = np.zeros(shape=(max_episodes, total_states), dtype=np.float32)
    policy = np.empty(total_states)
    Qs = np.zeros(shape=(max_episodes, total_states, total_actions), dtype=np.float32)
    policy_success = np.zeros(max_episodes, dtype=np.float32)
    return_reward = np.zeros(max_episodes, dtype=np.float32)

    # Double Q Learning
    for e in range(max_episodes):
      alpha = alphadecay[e]
      epsilon = epsilondecay[e]
      obs, info = env.reset()
      # mapping state to index (obs returns dict of agent storing current state and list of terminals)
      s = base_env.states_to_index[obs['agent']]
      done = False
      truncated = False
      '''
      Double Q-Learning uses two Q's instead of one to solve maximization bias problem as mentioned below:
      a = argmax((Q1[s] + Q2[s]) / 2)
      1. Q1(s,a) = Q1(s,a) + alpha * [ r + gamma * Q2(s',argmax(Q1[s'])) - Q1(s,a) ]
         where:
         td_target = r + gamma * Q2(s',argmax(Q1[s']))
         td_error = td_target - Q1(s,a)

      2.  Q2(s,a) = Q2(s,a) + alpha * [ r + gamma * Q1(s',argmax(Q2[s'])) - Q2(s,a) ]
         where:
         td_target = r + gamma * Q1(s',argmax(Q2[s']))
         td_error = td_target - Q2(s,a)

      Here, Step1 and Step2 are chosen randomly.

      '''
      while not done and not truncated:
        a = actionSelect(env, s, Q, epsilon)
        obs_prime, r, done, truncated, info = env.step(a)
        s_prime = base_env.states_to_index[obs_prime['agent']]

        if env.np_random.integers(2):
          a_q1 = np.argmax(Q1[s_prime])

          td_target = r
          if not done:
            td_target += gamma * Q2[s_prime][a_q1]
          td_error = td_target - Q1[s][a]

          Q1[s][a] += alpha * td_error

        else:
          a_q2 = np.argmax(Q2[s_prime])

          td_target = r
          if not done:
            td_target += gamma * Q1[s_prime][a_q2]
          td_error = td_target - Q2[s][a]
          Q2[s][a] += alpha * td_error

        s = s_prime

      Q = (Q1 + Q2) / 2
      Qs[e] = Q.copy()
      V_s[e] = np.max(Q, axis=1)

      # Evaluating Performance Metrics in each episode
      policy = np.argmax(Q, axis=1)
      for _ in range(config['rollouts']):
        # Finding policySucess and returnReward for each rollout using `get_policy_sucess(env,policy,config)`
        # See Section: get_policy_success(env, policy, config)
        policySuccess, returnReward = get_policy_success(env, policy, config)

        # adding policySucess of all rollouts within each episode
        policy_success[e] += policySuccess
        # adding returnReward of all rollouts within each episode
        return_reward[e] += returnReward

      # Taking average of policy_success and return_reward for number of rollouts used
      policy_success[e] /= config['rollouts']
      return_reward[e] /= config['rollouts']

    # Finally returning the mentioned variables above in docstring
    return Q, V_s, policy, Qs, None, policy_success, return_reward
```

```python
def plot_policy_success(policy_success, title, image_name, config=None):
    """
    Plots the evolution of Policy Success Rate (%) versus Episodes for multiple
    reinforcement learning algorithms.

    This function computes the mean policy success rate across runs for each
    algorithm and plots them on a single graph. Policy Success Rate is defined
    as the percentage of episodes in which the agent reaches the goal state
    when following the learned policy at that episode.

    The plot includes curves for:
        - First-Visit Monte Carlo Control (FVMC)
        - SARSA
        - Q-Learning
        - Double Q-Learning

    Args:
        policy_success (np.ndarray): Boolean array indicating policy success
            across episodes for each algorithm. Expected shape:
            (num_algorithms, max_episodes, num_runs) or equivalent.
        title (str): Title of the plot.
        image_name (str): Filename for saving the generated plot.
        config (dict): Configuration dictionary containing:
            - max_episodes (int): Number of training episodes.

    Returns:
        None: The function saves the plot to disk.
    """

    shape = policy_success.shape
    num_algorithms = shape[0]
    num_runs = shape[1]
    max_episodes = shape[2]

    label_to_algo = ['FVMC', 'SARSA', 'Q-Learning', 'Double-Q-Learning']
    plt.figure(figsize=(10, 6))

    for i in range(num_algorithms):

        # Mean success over runs -> shape (episodes,)
        success_rate = np.mean(policy_success[i], axis=0) * 100

        plt.plot(
            range(max_episodes),
            success_rate,
            linewidth=2,
            label=label_to_algo[i]
        )

    plt.xlabel("Episodes")
    plt.ylabel("Policy Success Rate (%)")
    plt.title(title)
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.savefig(image_name, dpi=300)
    plt.show()
    plt.close()
```

```python
def sarsa_lambda_control(environment, config=None):
    """
    Implements the SARSA(λ) algorithm with eligibility traces for a discrete
    Random Maze Environment (RME).

    This function applies SARSA(λ), an on-policy Temporal Difference (TD) control
    algorithm that extends SARSA by incorporating eligibility traces to speed up
    learning. Eligibility traces allow credit assignment to multiple previously
    visited state-action pairs. Both accumulating and replacing traces are
    supported, as specified in the configuration.

    Args:
        environment (gym.Env): Environment with discrete observation and action spaces.
        config (dict): Configuration dictionary containing:
            - max_episodes (int): Number of training episodes.
            - discount_factor (float): Discount factor γ.
            - lambda (float): Eligibility trace decay parameter λ.
            - trace (str): Type of eligibility trace ('accumulating' or 'replacing').
            - step_size (dict): Learning rate (alpha) decay parameters.
            - epsilon (dict): Epsilon-greedy exploration decay parameters.
            - seed (int): Random seed for environment reset.

    Returns:
        Q (np.ndarray): Final learned action-value function of shape (S, A).
        V_s (np.ndarray): State-value estimates derived from Q for each episode.
        policy (np.ndarray): Final greedy policy derived from Q.
        Qs (np.ndarray): History of Q-values across episodes.
        td_target (float): Last computed TD target value.
        policy_success (np.ndarray): Boolean success indicator per episode.
        return_reward (np.ndarray): Total return obtained per episode.
    """
    # Resetting the seed for reproducibility
    env = environment
    np.random.seed(config['seed'])
    random.seed(config['seed'])
    env.reset(seed=config['seed'])

    # Unwrapping the environment for mapping either states to index ot index to states
    # (Since I haven't register the wall state in the observation space so handling the mapping explicitly)
    base_env = env.unwrapped

    # Extracting required data from config
    gamma = config['discount_factor']
    max_episode = config['max_episodes']
    total_states = config['environment']['total_states']
    total_actions = config['environment']['total_actions']
    trace = config['trace']
    lambda_ = config['lambda']

    # Generating the episode-wise step sizes for 'max_episodes' number of episodes
    # See Section: generate_episode_step_sizes (above)
    alphadecay = generate_episode_step_sizes(**config['step_size'], max_episode = max_episode)

    # Generating the episode-wise epsilons for 'max_episodes' number of episodes
    # See Section: generate_episode_epsilons (above)
    epsilondecay = generate_episode_step_sizes(**config['epsilon'], max_episode=max_episode)

    # defining variables to be return in the end as mentioned above in docstring
    Q = np.zeros(shape=(total_states, total_actions), dtype=np.float32)
    Qs = np.zeros(shape=(max_episode, total_states, total_actions), dtype=np.float32)
    V_s = np.zeros(shape=(max_episode, total_states), dtype=np.float32)
    policy_success = np.zeros(shape=(max_episode), dtype=np.float32)
    return_reward = np.zeros(shape=(max_episode), dtype=np.float32)
    td_target = 0.0

    # SARSA(lambda) logic
    for e in range(max_episode):
      E = np.zeros(shape=(total_states, total_actions))
      alpha = alphadecay[e]
      epsilon = epsilondecay[e]
      obs, info = env.reset()
      s = base_env.states_to_index[obs['agent']]
      action = actionSelect(env, s, Q, epsilon)

      truncated = False
      done = False
      '''
      SARSA(lambda) replacing is an improvement over SARSA using Eligibility Traces as mentioned below:
      E[s,a] += 1
      clip(E, 0, 1)
      Q(s,a) = Q(s,a) + alpha * [ r + gamma * Q(s',a') - Q(s,a) ] * E
      E = gamma * lambda * E

      where:
      td_target = r + gamma * Q(s',a') and
      td_error = td_target - Q(s,a)
      '''
      while not done and not truncated:
        obs_prime, reward, done, truncated, info = env.step(action)
        s_prime = base_env.states_to_index[obs_prime['agent']]
        action_prime = actionSelect(env, s_prime, Q, epsilon)

        td_target = reward
        if not done:
          td_target += gamma * Q[s_prime, action_prime]

        td_error = td_target - Q[s, action]
        E[s, action] += 1

        if trace == "replacing":
          E = np.clip(E, 0, 1)

        Q += alpha * td_error * E
        E *= gamma * lambda_
        s = s_prime
        action = action_prime

      Qs[e] = Q.copy()
      V_s[e] = np.max(Q, axis=1)

      # Evaluating Performance Metrics in each episode
      policy = np.argmax(Q, axis=1)
      for _ in range(config['rollouts']):
        # Finding policySucess and returnReward for each rollout using `get_policy_sucess(env,policy,config)`
        # See Section: get_policy_success(env, policy, config)
        policySuccess, returnReward = get_policy_success(env, policy, config)

        # adding policySucess of all rollouts within each episode
        policy_success[e] += policySuccess
        # adding returnReward of all rollouts within each episode
        return_reward[e] += returnReward

      # Taking average of policy_success and return_reward for number of rollouts used
      policy_success[e] /= config['rollouts']
      return_reward[e] /= config['rollouts']

    # Finally returning the mentioned variables above in docstring
    return Q, V_s, policy, Qs, td_target, policy_success, return_reward
```

```python
def q_learning_lambda_control(environment, config=None):
    """
    Implements the Q(λ) algorithm with eligibility traces for a discrete
    Random Maze Environment (RME).

    This function applies Q(λ), an off-policy Temporal Difference (TD) control
    algorithm that extends Q-Learning by incorporating eligibility traces.
    Replacing or accumulating eligibility traces can be used to propagate
    temporal-difference errors across multiple previously visited state-action
    pairs. Trace resetting is handled according to whether the behavior policy
    follows the greedy action.

    Args:
        environment (gym.Env): Environment with discrete observation and action spaces.
        config (dict): Configuration dictionary containing:
            - max_episodes (int): Number of training episodes.
            - discount_factor (float): Discount factor γ.
            - lambda (float): Eligibility trace decay parameter λ.
            - trace (str): Type of eligibility trace ('accumulating' or 'replacing').
            - step_size (dict): Learning rate (alpha) decay parameters.
            - epsilon (dict): Epsilon-greedy exploration decay parameters.
            - seed (int): Random seed for environment reset.

    Returns:
        Q (np.ndarray): Final learned action-value function of shape (S, A).
        V_s (np.ndarray): State-value estimates derived from Q for each episode.
        policy (np.ndarray): Final greedy policy derived from Q.
        Qs (np.ndarray): History of Q-values across episodes.
        td_target (float): Last computed TD target value.
        policy_success (np.ndarray): Boolean success indicator per episode.
        return_reward (np.ndarray): Total return obtained per episode.
    """

    # Resetting the seed for reproducibility
    env = environment
    np.random.seed(config['seed'])
    random.seed(config['seed'])
    env.reset(seed=config['seed'])

    # Unwrapping the environment for mapping either states to index ot index to states
    # (Since I haven't register the wall state in the observation space so handling the mapping explicitly)
    base_env = env.unwrapped

    # Extracting required data from config
    max_episodes = config['max_episodes']
    gamma = config['discount_factor']
    total_states = config['environment']['total_states']
    total_actions = config['environment']['total_actions']
    trace = config['trace']
    lambda_ = config['lambda']

    # Generating the episode-wise step sizes for 'max_episodes' number of episodes
    # See Section: generate_episode_step_sizes (above)
    alphadecay = generate_episode_step_sizes(**config['step_size'], max_episode=max_episodes)

    # Generating the episode-wise epsilons for 'max_episodes' number of episodes
    # See Section: generate_episode_epsilons (above)
    epsilondecay = generate_episode_epsilons(**config['epsilon'], max_episode=max_episodes)

    # defining variables to be return in the end as mentioned above in docstring
    Q = np.zeros(shape=(total_states, total_actions), dtype=np.float32)
    V_s = np.zeros(shape=(max_episodes, total_states), dtype=np.float32)
    policy = np.empty(total_states)
    Qs = np.zeros(shape=(max_episodes, total_states, total_actions))
    td_target = 0.0
    policy_success = np.zeros(max_episodes, dtype=np.float32)
    return_reward = np.zeros(max_episodes, dtype=np.float32)

    is_action_greedy = False
    # Q(lambda) logic
    for e in range(max_episodes):
      E = np.zeros(shape=(total_states, total_actions), dtype=np.float32)
      alpha = alphadecay[e]
      epsilon = epsilondecay[e]
      obs, info = env.reset()
      s = base_env.states_to_index[obs['agent']]

      a = actionSelect(env, s, Q, epsilon)

      truncated = False
      done = False
      '''
      Q-Learning(lambda) replacing is an extension of Q-Learning that uses lambda-return for policy evaluation in GPI as mentioned below:
      E[s,:] = 0
      E[s,a] += 1
      Q(s,a) = Q(s,a) + alpha * [ r + gamma * max_a'(Q(s',a')) - Q(s,a) ] * E
      E = gamma * lambda * E if a' == argmax(Q(s')) else 0

      where:
      td_target = r + gamma * max_a'(Q(s',a'))
      td_error = td_target - Q(s,a)
      '''
      while not done and not truncated:
        obs_prime, r, done, truncated, info = env.step(a)
        s_prime = base_env.states_to_index[obs_prime['agent']]

        a_prime = actionSelect(env, s_prime, Q, epsilon)

        if a_prime == np.argmax(Q[s_prime]):
          is_action_greedy = True
        else:
          is_action_greedy = False

        td_target = r
        if not done:
          td_target += gamma * np.max(Q[s_prime])
        td_error = td_target - Q[s][a]

        if trace == 'replacing':
          E[s, :] = 0
        E[s, a] += 1

        Q += alpha * td_error * E
        if is_action_greedy:
          E *= gamma * lambda_
        else:
          E.fill(0)

        s = s_prime
        a = a_prime

      Qs[e] = Q.copy()
      V_s[e] = np.max(Q, axis=1)

      # Evaluating Performance Metrics in each episode
      policy = np.argmax(Q, axis=1)
      for _ in range(config['rollouts']):
        # Finding policySucess and returnReward for each rollout using `get_policy_sucess(env,policy,config)`
        # See Section: get_policy_success(env, policy, config)
        policySuccess, returnReward = get_policy_success(env, policy, config)

        # adding policySucess of all rollouts within each episode
        policy_success[e] += policySuccess
        # adding returnReward of all rollouts within each episode
        return_reward[e] += returnReward

      # Taking average of policy_success and return_reward for number of rollouts used
      policy_success[e] /= config['rollouts']
      return_reward[e] /= config['rollouts']

    # Finally returning the mentioned variables above in docstring
    return Q, V_s, policy, Qs, td_target, policy_success, return_reward
```

Dyna-Q(env, $\gamma$, $\alpha_0$, $\epsilon_0$, noEpisodes, noPlanning)

```python
def dyna_q_control(environment, config=None):
    """
    Implements the Dyna-Q algorithm for a discrete Random Maze Environment (RME).

    This function applies Dyna-Q, a model-based reinforcement learning algorithm
    that integrates direct learning from real environment interactions with
    planning updates using a learned model of the environment. The agent learns
    an action-value function Q(s, a) while simultaneously estimating the
    transition dynamics and reward model, which are used for simulated planning
    updates.

    Args:
        environment (gym.Env): Environment with discrete observation and action spaces.
        config (dict): Configuration dictionary containing:
            - max_episodes (int): Number of training episodes.
            - discount_factor (float): Discount factor γ.
            - no_planning_steps (int): Number of planning updates per real step.
            - step_size (dict): Learning rate (alpha) decay parameters.
            - epsilon (dict): Epsilon-greedy exploration decay parameters.
            - seed (int): Random seed for environment reset.

    Returns:
        Q (np.ndarray): Final learned action-value function of shape (S, A).
        V_s (np.ndarray): State-value estimates derived from Q for each episode.
        policy (np.ndarray): Final greedy policy derived from Q.
        Qs (np.ndarray): History of Q-values across episodes.
        None: Placeholder (no TD target history stored).
        policy_success (np.ndarray): Boolean success indicator per episode.
        return_reward (np.ndarray): Total return obtained per episode.
    """
    # reseting the seed for reproducibility
    env = environment
    np.random.seed(config['seed'])
    random.seed(config['seed'])
    env.reset(seed=config['seed'])

    # Unwrapping the environment for mapping either states to index ot index to states
    # (Since I haven't register the wall state in the observation space so handling the mapping explicitly)
    base_env = env.unwrapped

    # Extracting required data from config
    gamma = config['discount_factor']
    max_episode = config['max_episodes']
    total_states = config['environment']['total_states']
    total_actions = config['environment']['total_actions']

    # Generating the episode-wise step sizes for 'max_episodes' number of episodes
    # See Section: generate_episode_step_sizes (above)
    alphadecay = generate_episode_step_sizes(**config['step_size'], max_episode=max_episode)

    # Generating the episode-wise epsilons for 'max_episodes' number of episodes
    # See Section: generate_episode_epsilons (above)
    epsilondecay = generate_episode_epsilons(**config['epsilon'], max_episode=max_episode)

    # defining variables to be return in the end as mentioned above in docstring
    Q = np.zeros(shape=(total_states, total_actions), dtype=np.float32)
    V_s = np.zeros(shape=(max_episode, total_states), dtype=np.float32)
    policy = np.empty(total_states)
    Qs = np.zeros(shape=(max_episode, total_states, total_actions))
    policy_success = np.zeros(max_episode, dtype=np.float32)
    return_reward = np.zeros(max_episode, dtype=np.float32)

    T = np.zeros(shape=(total_states, total_actions, total_states), dtype=np.float32)
    R = np.zeros(shape=(total_states, total_actions, total_states), dtype=np.float32)

    # Dyna-Q logic
    for e in range(max_episode):
      alpha = alphadecay[e]
      epsilon = epsilondecay[e]
      E = np.zeros(shape=(total_states, total_actions), dtype=np.float32)
      obs, info = env.reset()
      s = base_env.states_to_index[obs['agent']]
      done = False
      truncated = False
      '''
      Dyna-Q is a model based learning which updates the Q multiple times in each episode as mentioned below:
      Q(s,a) = Q(s,a) + alpha * [ r + gamma * max_a'(Q(s',a')) ]
      T[s,a,s'] += 1
      R[s,a,s'] += (r - R[s,a,s']) / T[s,a,s']

      Repeat for no_planning_steps:
        s_p ~ T & a_p ~ T
        s'_p ~ S with probability p = T[s_p,a_p] / sum(T[s_p,a_p,:])
        r = R[s_p,a_p,s'_p]
        Q(s_p,a_p) = Q(s_p,a_p) + alpha * [ r + gamma * max_a'(Q(s'_p,a')) ]

      where:
        td_target = r + gamma * max_a'(Q(s',a'))
        td_error = td_target - Q(s,a)
      '''
      while not done and not truncated:
        a = actionSelect(env, s, Q, epsilon)
        obs_prime, r, done, truncated, info = env.step(a)
        s_prime = base_env.states_to_index[obs_prime['agent']]
        T[s,a,s_prime] += 1
        rdiff = r - R[s, a, s_prime]
        R[s, a, s_prime] += rdiff / T[s, a, s_prime]

        td_target = r

        if not done:
          td_target += gamma * np.max(Q[s_prime])
        td_error = td_target - Q[s][a]

        Q[s,a] += alpha * td_error
        s_backup = s_prime

        for _ in range(config['no_planning_steps']):
          if np.sum(Q) == 0:
            continue
          s = env.np_random.choice(total_states)
          a = env.np_random.choice(total_actions)
          if np.sum(T[s,a,:]) == 0:
            continue
          prob_s_prime = T[s,a] / np.sum(T[s,a,:])
          s_prime = np.random.choice(total_states, p=prob_s_prime)
          r = R[s,a,s_prime]
          td_target = r + gamma * np.max(Q[s_prime])
          td_error = td_target - Q[s,a]
          Q[s,a] += alpha * td_error

        s = s_backup

      Qs[e] = Q.copy()
      V_s[e] = np.max(Q, axis=1)

      # Evaluating Performance Metrics in each episode
      policy = np.argmax(Q, axis=1)
      for _ in range(config['rollouts']):
        # Finding policySucess and returnReward for each rollout using `get_policy_sucess(env,policy,config)`
        # See Section: get_policy_success(env, policy, config)
        policySuccess, returnReward = get_policy_success(env, policy, config)

        # adding policySucess of all rollouts within each episode
        policy_success[e] += policySuccess
        # adding returnReward of all rollouts within each episode
        return_reward[e] += returnReward

      # Taking average of policy_success and return_reward for number of rollouts used
      policy_success[e] /= config['rollouts']
      return_reward[e] /= config['rollouts']

    # Finally returning the mentioned variables above in docstring
    return Q, V_s, policy, Qs, None, policy_success, return_reward
```

# **Problem 11: Trajectory Learning**

Implement the Trajectory Learning algorithm for the Random Maze Environment (RME) described above.
Use the function definition as given in Lecture slides.

TrajectorySampling(env, γ, α0, ϵ0, noEpisodes, maxTrajectory)

```python
def trajectory_sampling_control(environment, config=None):
    """
    Trajectory Sampling (Trajectory Learning) Control Algorithm

    Algorithm (from lecture slides):
        TrajectorySampling(env, γ, α0, ε0, noEpisodes, maxTrajectory)

    Goal:
        Learn an optimal action-value function Q(s, a) by combining:
            - Direct interaction with the environment
            - A learned model of the environment
            - Simulated trajectories sampled from the learned model

    The algorithm alternates between:
        1. Real experience updates (Q-learning)
        2. Planning updates via trajectory sampling

    Inputs:
        environment : Random Maze Environment (RME)
        config      : Dictionary containing γ, α0, ε0, number of episodes,
                      maximum trajectory length, and sampling probability
    """
    # reseting the seed for reproducibility
    env = environment
    np.random.seed(config['seed'])
    random.seed(config['seed'])
    env.reset(seed=config['seed'])

    # Unwrapping the environment for mapping either states to index ot index to states
    # (Since I haven't register the wall state in the observation space so handling the mapping explicitly)
    base_env = env.unwrapped

    # Extracting required data from config
    gamma = config['discount_factor']
    max_episode = config['max_episodes']
    total_states = config['environment']['total_states']
    total_actions = config['environment']['total_actions']

    # Generating the episode-wise step sizes for 'max_episodes' number of episodes
    # See Section: generate_episode_step_sizes (above)
    alphadecay = generate_episode_step_sizes(**config['step_size'], max_episode=max_episode)

    # Generating the episode-wise epsilons for 'max_episodes' number of episodes
    # See Section: generate_episode_epsilons (above)
    epsilondecay = generate_episode_epsilons(**config['epsilon'], max_episode=max_episode)

    # defining variables to be return in the end as mentioned above in docstring
    Q = np.zeros(shape=(total_states, total_actions), dtype=np.float32)
    V_s = np.zeros(shape=(max_episode, total_states), dtype=np.float32)
    policy = np.empty(total_states)
    Qs = np.zeros(shape=(max_episode, total_states, total_actions))
    policy_success = np.zeros(max_episode, dtype=bool)
    return_reward = np.zeros(max_episode, dtype=np.float32)

    T = np.zeros(shape=(total_states, total_actions, total_states), dtype=np.float32)
    R = np.zeros(shape=(total_states, total_actions, total_states), dtype=np.float32)

    # Trajectory Learning Logic
    for e in range(max_episode):
      alpha = alphadecay[e]
      epsilon = epsilondecay[e]
      E = np.zeros(shape=(total_states, total_actions), dtype=np.float32)
      obs, info = env.reset()
      s = base_env.states_to_index[obs['agent']]
      done = False
      truncated = False
      '''
      Trajectory Learning is also a model based learning and very simimlar to Dyna-Q.
      Only difference is that here state s_p is not sampled from T, instead we use initial state s.

      Q(s,a) = Q(s,a) + alpha * [ r + gamma * max_a'(Q(s',a')) ]
      T[s,a,s'] += 1
      R[s,a,s'] += (r - R[s,a,s']) / T[s,a,s']

      Repeat for max_trajectory:
        s_p = s
        action a_p is chosen either greedily or randomly with probability epsilon
        s'_p ~ S with probability p = T[s_p,a_p] / sum(T[s_p,a_p,:])
        r = R[s_p,a_p,s'_p]
        Q(s_p,a_p) = Q(s_p,a_p) + alpha * [ r + gamma * max_a'(Q(s'_p,a')) ]

      where:
        td_target = r + gamma * max_a'(Q(s',a'))
        td_error = td_target - Q(s,a)
      '''
      while not done and not truncated:
        a = actionSelect(env, s, Q, epsilon)
        obs_prime, r, done, truncated, info = env.step(a)
        s_prime = base_env.states_to_index[obs_prime['agent']]
        T[s,a,s_prime] += 1
        rdiff = r - R[s, a, s_prime]
        R[s, a, s_prime] += rdiff / T[s, a, s_prime]

        td_target = r

        if not done:
          td_target += gamma * np.max(Q[s_prime])
        td_error = td_target - Q[s][a]

        Q[s,a] += alpha * td_error
        s_backup = s_prime

        for _ in range(config['maxTrajectory']):
          if np.sum(Q) == 0:
            break
          a = actionSelect(env, s, Q, epsilon)
          if np.sum(T[s,a,:]) == 0:
            break
          prob_s_prime = T[s,a] / np.sum(T[s,a,:])
          s_prime = np.random.choice(total_states, p=prob_s_prime)
          r = R[s,a,s_prime]
          td_target = r + gamma * np.max(Q[s_prime])
          td_error = td_target - Q[s,a]
          Q[s,a] += alpha * td_error
          s = s_prime

        s = s_backup

      Qs[e] = Q.copy()
      V_s[e] = np.max(Q, axis=1)

      # Evaluating Performance Metrics in each episode
      policy = np.argmax(Q, axis=1)
      for _ in range(config['rollouts']):
        # Finding policySucess and returnReward for each rollout using `get_policy_sucess(env,policy,config)`
        # See Section: get_policy_success(env, policy, config)
        policySuccess, returnReward = get_policy_success(env, policy, config)

        # adding policySucess of all rollouts within each episode
        policy_success[e] += policySuccess
        # adding returnReward of all rollouts within each episode
        return_reward[e] += returnReward

      # Taking average of policy_success and return_reward for number of rollouts used
      policy_success[e] /= config['rollouts']
      return_reward[e] /= config['rollouts']

    # Finally returning the mentioned variables above in docstring
    return Q, V_s, policy, Qs, None, policy_success, return_reward
```

```python
def plot_policy_success(policy_success, title, image_name, config=None):
    """
    Plot Policy Success Rate (%) vs Episodes for multiple algorithms.

    Task:
        For each algorithm, plot the evolution of policy success rate
        (in percentage) as training progresses.

    Definition:
        Policy Success Rate =
            (Number of times the agent reaches the goal state) /
            (Total number of evaluation episodes)

    Inputs:
        policy_success : Array containing policy success outcomes across
                         episodes and multiple runs for each algorithm
        title          : Title of the plot
        image_name     : Filename to save the plot
        config         : Configuration dictionary containing max_episodes
    """
    shape = policy_success.shape
    num_algorithms = shape[0]
    num_runs = shape[1]
    max_episodes = shape[2]

    label_to_algo = ['SARSA(λ) Replacing', 'SARSA(λ) Accumulating', 'Q(λ) Replacing', 'Q(λ) Accumulating', 'Dyna-Q', 'Trajectory Learning']
    plt.figure(figsize=(10, 6))

    for i in range(num_algorithms):

        # Mean success over runs -> shape (episodes,)
        success_rate = np.mean(policy_success[i], axis=0) * 100

        plt.plot(
            range(max_episodes),
            success_rate,
            linewidth=2,
            label=label_to_algo[i]
        )

    plt.xlabel("Episodes")
    plt.ylabel("Policy Success Rate (%)")
    plt.title(title)
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.savefig(image_name, dpi=300)
    plt.show()
    plt.close()
```

###(e) State-value Function Estimation Error (5 Points)
For each of the algorithm (in a single plot), plot the State-value Function Estimation Error vs Episodes.
State-value Function Estimation Error is defined as Mean Absolute Error across all V-function estimates
(across all states) from the respective optimal value.

```python
def plot_state_value_error(state_values, Vopt, title, image_name, config=None):
    """
    Plot State-value Function Estimation Error vs Episodes
    for multiple reinforcement learning algorithms.

    Task:
        For each algorithm, plot how the error in the estimated
        state-value function decreases over training episodes.

    Definition:
        State-value Function Estimation Error =
            Mean Absolute Error (MAE) between the estimated V(s)
            and the true optimal V*(s), averaged across all
            non-terminal states.

    Inputs:
        state_values : Array containing estimated V-values for all
                       states, episodes, and algorithms
        title        : Title of the plot
        image_name   : Filename to save the plot
        config       : Configuration dictionary containing max_episodes
    """
    shape = state_values.shape
    num_algorithms = shape[0]
    num_runs = shape[1]
    max_episodes = shape[2]
    num_states = shape[3]

    label_to_algo = ['SARSA(λ) Replacing', 'SARSA(λ) Accumulating', 'Q(λ) Replacing', 'Q(λ) Accumulating', 'Dyna-Q', 'Trajectory Learning']
    plt.figure(figsize=(10, 6))

    for i in range(num_algorithms):
        # Average over runs -> shape (episodes, states)
        mean_V = np.mean(state_values[i], axis=0)

        # MAE per episode -> shape (episodes,)
        mae = np.mean(np.abs(mean_V - Vopt), axis=1)

        plt.plot(range(max_episodes), mae, linewidth=2, label=label_to_algo[i])

    plt.xlabel("Episodes")
    plt.ylabel("State-Value Function Estimation Error")
    plt.title(title)
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.savefig(image_name, dpi=300)
    plt.show()
    plt.close()
```
