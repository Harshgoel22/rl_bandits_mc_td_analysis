## OpenAI Gym Environments


### Random-Maze Environment

In this assignment we will be exploring a variant of the **Random Maze Environment (RME)** that we have been looking at in the lectures.

#### Environment Description

The environment is represented as a **grid world** (Figure a). Random maze environment is a **highly stochastic environment** with:

- **11 states**: Two terminal states (a goal state **G** and a hole state **H**) and 9 non-terminal states
- **A wall** in between the environment

The **wall behaves similar to the wall on the periphery** of the environment: if an agent bumps against the wall, it **bounces back**. The boundary of the environment behaves similarly.

#### Grid Layout (3×4):

```
[ 0 ][ 1 ][ 2 ][ 3-G ]
[ 4 ][  5-WALL  ][ 6 ][ 7-H ]
[ 8-S ][ 9 ][ 10][ 11]
```

Where:
- **S** (State 8): Start state
- **G** (State 3): Goal state
- **H** (State 7): Hole state
- **WALL** (State 5): Wall state (agent bounces back)

#### Rewards:

- Agent receives a reward of **+1** when it lands in the **goal state (3)**
- Agent receives a reward of **-1** when it lands in the **hole state (7)**
- For rest of the transitions, there is a reward of **-0.04** (living cost)

#### Transitions:

The transitions are **stochastic** as shown in Figure b. In this environment, **four actions** are possible: **left (2)**, **up (1)**, **right (0)**, and **down (3)**.

For every intended action:
- There is **80% chance** of going in the intended direction
- Remaining **20% chance** of going in either of the orthogonal directions
- The 20% chance gets **equally distributed** between each of the orthogonal directions (10% each)

The agent starts from **state 8 (S)**.

#### Discount Factor:

Assume **γ = 0.99** for all problems below.

````python
class SlipperyMazeEnv(Env):
    """
    A stochastic grid-based maze environment where actions may result
    in unintended movements due to slippery transitions.

    CRITICAL: State 5 is a WALL - agent bounces back if it tries to move there!
    """

    def __init__(self, config=None):
        """
        Initialize the Slippery Maze Environment.

        Args:
            config: Dictionary containing environment configuration
        """
        super().__init__()
        self.config = config['environment'] or {}
        self.terminal_states = np.array([self.config['goal_state'], self.config['hole_state']], dtype=np.int32)
        self.GOAL_STATE = self.config['goal_state']
        self.WALL_STATE = self.config['wall_state']
        self.HOLE_STATE = self.config['hole_state']
        self.START_STATE = self.config['start_state']
        self.agent_location = self.config['start_state']
        self.observation_space = spaces.Dict({
            'agent': spaces.Discrete(self.config['total_states'] + 1), # +1 for including wall in the total_states for easier handling
            'terminals': spaces.Box(low=0, high=self.config['total_states']-1, shape=(2,), dtype=np.int32)
        })
        self.action_space = spaces.Discrete(self.config['total_actions'])
        self.P = self.make_environment(self.config)


    # created this method for my ease to push the states, can do within make_environment() but reducing the readability there
    def _push_state_after_checks(self, P, s, action, ni, nj, prob):
      (x, y) = self.config['maze_shape']
      s_prime = ni * y + nj

      if ni < 0 or ni >= x or nj < 0 or nj >= y or s_prime == self.WALL_STATE:
        rwd = self.config['living_reward']
        P[s][action].append((prob, s, rwd, False))

      else:
        rwd = self.get_reward(s_prime)
        done = self.is_terminated(s_prime)
        P[s][action].append((prob, s_prime, rwd, done))

    def make_environment(self, config):
        """
        Creates the transition probability model P[s][a] for the maze.

        Grid Layout (3x4):
        [ 0][ 1][ 2][ 3-G]
        [ 4][  5-WALL  ][ 6][ 7-H]
        [ 8-S][ 9][10][11]

        Actions: 0=right, 1=up, 2=left, 3=down

        Transition Probabilities:
        - 80% chance of moving in intended direction
        - 10% chance of moving in each orthogonal direction
        - 0% chance of moving in opposite direction

        Returns:
            P: Dictionary where P[state][action] = [(prob, next_state, reward, done), ...]
        """
        P = defaultdict(lambda : defaultdict(list))
        (x, y) = config['maze_shape']
        intended_action_prob = 0.8
        orthogonal_action_prob = 0.1

        # 0 -> right | 1 -> up | 2 -> left | 3 -> down
        directions = {0: (0,1), 1: (-1,0), 2: (0,-1), 3: (1,0)}
        orthogonal_directions = {0: [1,3], 1: [0,2], 2: [1,3], 3: [0,2]}

        for i in range(x):
          for j in range(y):
            curr_state = i * y + j
            if(curr_state == self.WALL_STATE):
              continue

            # for intended action
            for action in range(self.action_space.n):
              # checking whether curr_state is terminal state
              if curr_state == self.GOAL_STATE or curr_state == self.HOLE_STATE:
                P[curr_state][action].append((1.0, curr_state, 0.0, True))
                continue

              # runs only for non terminal states
              (dx, dy) = directions[action]
              ni = i + dx
              nj = j + dy
              self._push_state_after_checks(P, curr_state, action, ni, nj, intended_action_prob)

              ## for orthogonal actions
              for ortho_action in orthogonal_directions[action]:
                (ox, oy) = directions[ortho_action]
                oi = i + ox
                oj = j + oy
                self._push_state_after_checks(P, curr_state, action, oi, oj, orthogonal_action_prob)

        return P

    def is_terminated(self, state):
        """
        Check if the given state is a terminal state.

        Args:
            state: Current state

        Returns:
            bool: True if state is goal or hole, False otherwise
        """
        # TODO: Implement
        return True if state in self.terminal_states else False

    def get_reward(self, state):
        """
        Get the reward for transitioning to the given state.

        Args:
            state: State to get reward for

        Returns:
            float: +1 for goal, -1 for hole, -0.04 for others
        """
        # TODO: Implement
        return 1.0 if state == self.GOAL_STATE else -1.0 if state == self.HOLE_STATE else -0.04

    def get_obs(self):
        """Get current observation."""
        return {'agent': self.agent_location, 'terminals': self.terminal_states}

    def get_info(self):
        """Get additional information."""
        state_x = self.agent_location // self.config['maze_shape'][1]
        state_y = self.agent_location % self.config['maze_shape'][1]
        goal_x = self.GOAL_STATE // self.config['maze_shape'][1]
        goal_y = self.GOAL_STATE % self.config['maze_shape'][1]
        return {'distance': abs(state_x - goal_x) + abs(state_y - goal_y)}

    def reset(self, seed=None, options=None):
        """
        Reset the environment to the initial state.

        Args:
            seed: Random seed for reproducibility
            options: Additional options

        Returns:
            observation: Initial observation
            info: Additional information
        """
        super().reset(seed=seed)
        self.agent_location = self.config['start_state']
        obs = self.get_obs()
        info = self.get_info()
        return obs, info


    def step(self, action):
        """
        Execute one time step within the environment.

        Args:
            action: Action to take (0=right, 1=up, 2=left, 3=down)

        Returns:
            observation: New observation after taking action
            reward: Reward received
            terminated: Whether episode has terminated
            truncated: Whether episode was truncated
            info: Additional information
        """
        possible_transition_states = self.P[self.agent_location][action]
        next_state_probs = [prob for prob, _, _, _ in possible_transition_states]
        stateidx = self.np_random.choice((len(possible_transition_states)), p=next_state_probs)
        _, s_prime, rwd, done = possible_transition_states[stateidx]
        self.agent_location = s_prime
        obs = self.get_obs()
        info = self.get_info()
        return obs, rwd, done, False, info


    def render(self):
        """Render the environment (optional)."""
        gridx, gridy = self.config['maze_shape']

        fig, ax = plt.subplots(figsize=(gridy*1.4, gridx*1.4))
        ax.set_xlim(0, gridy)
        ax.set_ylim(0, gridx)
        ax.set_xticks([])
        ax.set_yticks([])
        ax.set_aspect('equal')
        ax.invert_yaxis()

        for i in range(gridx):
            for j in range(gridy):
                state = i * gridy + j

                color = "#000000" if state == env.unwrapped.WALL_STATE else "#023020" if state == env.unwrapped.GOAL_STATE else "#8B0000" if state == env.unwrapped.HOLE_STATE else "#f5f5f5"

                rect = Rectangle((j, i), 1, 1, facecolor=color, edgecolor="black", linewidth=1.8)
                ax.add_patch(rect)

                cx, cy = j + 0.5, i + 0.5
                ax.text(cx+0.38, cy+0.38, f'${state}', ha='center', va='center', fontsize=12, color="black")
                if state == self.agent_location:
                  ax.text(cx, cy-0.28, "A", ha='center', va='center', fontsize=18, weight='bold', color="xkcd:dark red")

                if state == self.START_STATE:
                  ax.text(cx+0.38, cy-0.38, "S", ha='center', va='center', fontsize=18, weight='bold', color="black")

                if state == self.WALL_STATE:
                    ax.text(cx+0.38, cy-0.38, "W", ha='center', va='center', fontsize=18, weight='bold', color="white")
                    ax.text(cx+0.38, cy+0.38, f'{state}', ha='center', va='center', fontsize=12, color="white")

                elif state == self.GOAL_STATE:
                    ax.text(cx+0.38, cy-0.38, "G", ha='center', va='center', fontsize=18, weight='bold', color="white")
                    ax.text(cx, cy, "+1", ha='center', va='center', fontsize=14, color="darkgreen")
                    ax.text(cx+0.38, cy+0.38, f'{state}', ha='center', va='center', fontsize=12, color="white")

                elif state == self.HOLE_STATE:
                    ax.text(cx+0.38, cy-0.38, "H", ha='center', va='center', fontsize=18, weight='bold', color="white")
                    ax.text(cx, cy, "-1", ha='center', va='center', fontsize=14, color="xkcd:light red")
                    ax.text(cx+0.38, cy+0.38, f'{state}', ha='center', va='center', fontsize=12, color="white")


        # Thick outer border
        for spine in ax.spines.values():
            spine.set_visible(True)
            spine.set_linewidth(2)

        plt.title("Maze Environment", fontsize=14, pad=10)
        fig.text(0.5, -0.02, "Living Penalty of -0.04", ha="center", fontsize=14, color="darkgreen", weight="bold")
        plt.tight_layout()
        plt.show()

    def close(self):
        """Close the environment (optional)."""
        del self.P
        del self.observation_space
        del self.action_space
        del self.config
        del self.terminal_states
        del self.GOAL_STATE
        del self.WALL_STATE
        del self.HOLE_STATE
        del self.agent_location
````

# Problem 2: RME Optimal Policy via Dynamic Programming (90 Marks)

## Introduction

In this problem, we **assume we know the underlying MDP** of the Random Maze Environment. We are interested in finding the **optimal policy** for this environment using **Dynamic Programming** methods.

Dynamic Programming methods are guaranteed to find the optimal policy when the MDP is known. We will implement and compare two classic DP algorithms:

1. **Policy Iteration (PI)**
2. **Value Iteration (VI)**

### Key Concepts:

- **Policy Evaluation:** Compute value function $V^π$ for a given policy $π$
- **Policy Improvement:** Generate better policy from value function
- **Value Iteration:** Combine evaluation and improvement in one step

## Part 1: Policy Iteration (40 Marks)

### Algorithm Overview:

Policy Iteration alternates between two steps:

1. **Policy Evaluation:** Given a policy $π$, compute $V^π(s)$  for all states
   - Iteratively solve: $V(s) \leftarrow \sum_a \pi(a|s) \sum_{s',~r} p(s',r|s,a)[r + \gamma V(s')]$

   - Stop when $max|V_{new}(s) - V_{old}(s)| < θ$

2. **Policy Improvement:** Update policy greedily w.r.t. current $V$

   - $π(s) ← argmax_{a}~Σ_{s',~r} p(s',r|s,a)[r + γV(s')]$
   
   - If policy doesn't change, we have found optimal policy

### Requirements:

1. **Implement `policy_evaluation(env, policy, gamma, theta, config)`**
   - Input: Environment, current policy, discount factor, threshold
   - Output: Value function $V$ for given policy
   - Must converge when $max~change < θ$

2. **Implement `policy_improvement(env, V, gamma, config)`**
   - Input: Environment, value function, discount factor
   - Output: Improved policy, boolean indicating if policy changed
   - Must select greedy action for each state

3. **Implement `policy_iteration(env, gamma, theta, config)`**
   - Input: Environment, discount factor, threshold
   - Output: Optimal policy, optimal value function, number of iterations
   - Must alternate between evaluation and improvement until convergence

### Deliverables:

1. **Code Implementation:** Complete working implementation of all three functions
2. **Initial Policy Description:** Clearly describe the random policy you started with
3. **Optimal Policy Visualization:** Draw/plot the final optimal policy obtained
4. **Convergence Analysis:** Report how many iterations required to converge
5. **Discussion:** Explain any interesting observations


***Mathematical Equation used in Policy Evaluation:***

$V[s] = ∑_{a∈𝔸} ℿ[s,a] ∑_{(p,s',r,done) ∈ 𝑃[s][a]} p (r + γV[s'])$  $∀s ∈ 𝐒$

***Mathematical Equation used in Policy Improvement:***

$Q[s][a] = ∑_{(p,s',r,done) ∈ 𝑃[s][a]} p (r + γV[s'])$  $∀s ∈ 𝐒 \& ∀a ∈ 𝔸$

$ℿ[s,argmax(Q[s]) = 1$  $ ∀s ∈ 𝐒$

````python
def policy_evaluation(env, policy, gamma, theta, config):
    """
    Evaluate a policy given an environment and a discount factor.

    This function computes the value function V^π(s) for all states under
    the given policy π using iterative policy evaluation.

    Args:
        env: The environment (with transition model env.P)
        policy: Current policy (numpy array of shape [num_states, num_actions])
                policy[s, a] = probability of taking action a in state s
        gamma: Discount factor (0.99)
        theta: Convergence threshold (1e-10)
        config: Configuration dictionary

    Returns:
        V: Value function (numpy array of shape [num_states])
           V[s] = expected return starting from state s following policy π
    """
    # TODO: Implement policy evaluation
    """
    Here, we are iteratively computing the Vnew until absolute difference between the
    Vnew and Vold is less than θ using the update rule mentioned below:

        V[s] = ∑_{a∈A} Π[s,a] ∑_{(p,s′,r,done)∈P[s][a]} p * (r + γ * V[s′])    ∀s∈S
        if max|Vnew(s) - Vold(s)| < θ:
          break
    """
    V = np.zeros(shape=env.observation_space['agent'].n, dtype=np.float32)
    while 1:
      Vnew = np.zeros(shape=env.observation_space['agent'].n, dtype=np.float32)
      for s in range(env.observation_space['agent'].n):
        if s == env.unwrapped.WALL_STATE:
          continue
        for a in range(env.action_space.n):
          temp = 0
          for p, s_prime, rwd, done in env.unwrapped.P[s][a]:
            temp += p * (rwd + gamma * V[s_prime])
          Vnew[s] += policy[s,a] * temp
      if np.max(abs(Vnew - V)) < theta:
        break
      V = Vnew
    return V

def policy_improvement(env, V, gamma, config):
    """
    Improve policy greedily with respect to the current value function.

    This function generates a better (or equally good) policy by acting
    greedily with respect to the current value function V.

    Args:
        env: The environment (with transition model env.P)
        V: Current value function (numpy array of shape [num_states])
        gamma: Discount factor (0.99)
        config: Configuration dictionary

    Returns:
        policy: Improved policy (numpy array of shape [num_states, num_actions])
                policy[s, a] = 1 if a = argmax Q(s,a), else 0
        policy_stable: Boolean indicating if policy changed
                      True = policy didn't change (converged)
                      False = policy changed (need another iteration)
    """
    # TODO: Implement policy improvement
    """
    Here, using the updated state-values, first we are computing the state-action values i.e. Q, then
    using it, finding the optimal policy using the update rule mentioned below:

        Q[s][a] = ∑_{(p,s',r,done) ∈ 𝑃[s][a]} p * (r + γ * V[s'])  ∀s ∈ 𝐒 & ∀a ∈ 𝔸

        ℿ[s,argmax(Q[s]) = 1   ∀s ∈ 𝐒
    """
    Q = np.zeros(shape=(env.observation_space['agent'].n, env.action_space.n), dtype=np.float32)
    policy = np.zeros(shape=(env.observation_space['agent'].n, env.action_space.n), dtype=np.int32)
    policy_stable = True

    for s in range(env.observation_space['agent'].n):
      if s == env.unwrapped.WALL_STATE:
        continue
      for a in range(env.action_space.n):
        for p, s_prime, rwd, _ in env.unwrapped.P[s][a]:
          Q[s,a] += p * (rwd + gamma * V[s_prime])

    for s in range(env.observation_space['agent'].n):
      if s == env.unwrapped.WALL_STATE:
        continue
      policy[s, np.argmax(Q[s])] = 1

    old_policy = config['policy']
    if not np.array_equal(old_policy, policy):
      policy_stable = False
    return policy, policy_stable



def policy_iteration(env, gamma, theta, config):
    """
    Find optimal policy using Policy Iteration algorithm.

    This function alternates between policy evaluation and policy improvement
    until the policy converges to the optimal policy π*.

    Args:
        env: The environment (with transition model env.P)
        gamma: Discount factor (0.99)
        theta: Convergence threshold for policy evaluation (1e-10)
        config: Configuration dictionary

    Returns:
        policy: Optimal policy (numpy array)
        V: Optimal value function (numpy array)
        iterations: Number of iterations until convergence

    """
    # TODO: Implement policy iteration

    """
    In Policy Iteration, we are running Policy Evaluation and Policy Improvement alternatively until the policy converges.
    """
    policy = config['policy']
    V = np.zeros(shape=env.observation_space['agent'].n, dtype=np.float32)
    time_per_iteration = []
    iteration = 0

    gs_time = time.time()
    while 1:
      iteration += 1
      config['policy'] = policy

      stime = time.time()
      V = policy_evaluation(env, policy, gamma, theta, config)
      policy, policy_stable = policy_improvement(env, V, gamma, config)
      time_per_iteration.append(time.time() - stime)

      if policy_stable:
        break
    gt_time = time.time() - gs_time

    config['time_per_iteration'] = np.mean(time_per_iteration)
    config['total_time'] = gt_time
    return policy, V, iteration
````

## Part 2: Value Iteration (40 Marks)

### Algorithm Overview:

Value Iteration combines policy evaluation and improvement into a single update:

- **Single Update:** $V(s) ← max_a ~Σ_{s',~r} p(s',r|s,a)~[r + γV(s')]$

- **Convergence:** Stop when $max~|V_{new}(s) - V_{old}(s)| < θ$

- **Extract Policy:** After convergence, extract greedy policy

### Key Difference from Policy Iteration:

- **PI:** Fully evaluate each policy before improving it
- **VI:** Update value function with max over actions directly (no explicit policy)

### Requirements:

1. **Implement `value_iteration(env, gamma, theta, config)`**
   - Input: Environment, discount factor, threshold
   - Output: Optimal value function, optimal policy, number of iterations
   - Must use Bellman optimality equation for updates

### Deliverables:

````python
def value_iteration(env, gamma, theta, config):
    """
    Find optimal policy using Value Iteration algorithm.

    This function iteratively updates the value function using the Bellman
    optimality equation until convergence, then extracts the optimal policy.

    Args:
        env: The environment (with transition model env.P)
        gamma: Discount factor (0.99)
        theta: Convergence threshold (1e-10)
        config: Configuration dictionary

    Returns:
        V: Optimal value function (numpy array)
        policy: Optimal policy (numpy array)
        iterations: Number of iterations until convergence
    """
    # TODO: Implement value iteration
    """
    In Value Iteration, we are iteratively computing the Vnew until absolute difference between the
    Vnew and Vold is less than θ.
    First we are computing the state-action values i.e. Q then using it, finding the state-values and
    optimal policy as per the update rule mentioned below:

            Q[s][a] = ∑_{(p,s',r,done) ∈ 𝑃[s][a]} p * (r + γ * V[s'])  ∀s ∈ 𝐒 & ∀a ∈ 𝔸
            V = max(Q[s])
            if max|Vnew(s) - Vold(s)| < θ:
              break
            ℿ[s,argmax(Q[s]) = 1   ∀s ∈ 𝐒
    """
    V = np.zeros(shape=env.observation_space['agent'].n, dtype=np.float32)
    policy = np.zeros(shape=(env.observation_space['agent'].n, env.action_space.n), dtype=np.int32)
    iterations = 0
    time_per_iteration = []

    gs_time = time.time()
    while 1:
      iterations += 1
      Q = np.zeros(shape=(env.observation_space['agent'].n, env.action_space.n), dtype=np.float32)

      stime = time.time()
      for s in range(env.observation_space['agent'].n):
        if s == env.unwrapped.WALL_STATE:
          continue
        for a in range(env.action_space.n):
          for p, s_prime, rwd, _ in env.unwrapped.P[s][a]:
            Q[s,a] += p * (rwd + gamma * V[s_prime])
      ttime = time.time() - stime
      time_per_iteration.append(ttime)

      if np.max(abs(V - np.max(Q, axis=1))) < theta:
        break
      V = np.max(Q, axis=1)

    for s in range(env.observation_space['agent'].n):
      if s == env.unwrapped.WALL_STATE:
        continue
      policy[s, np.argmax(Q[s])] = 1
    gt_time = time.time() - gs_time

    config['time_per_iteration'] = np.mean(time_per_iteration)
    config['total_time'] = gt_time

    return V, policy, iterations
````

## Utility Functions

***Mathematical Equations used in decay α:***
1. For linear α decay:
  
  >> $decay\_factor = (\alpha_{final} - \alpha_{initial}) / (max_steps - 1)$
  >>
  >> $α_{new} = α_{initial} + episode * decay\_factor$

2. For exponential α decay:

  >> $decay\_factor = \left(\frac{\alpha_{final}}{\alpha_{initial}}\right)^{\frac{1}{max\_steps - 1}}$
  >>
  >> $α_{new} = α_{old} + episode * decay\_factor$

***Mathematical Equation used in finding true values:***
  
  >> $V[s] = ∑_{a∈A} Π[s,a] ∑_{(p,s′,r,done)∈P[s][a]} p * (r + γ * V[s′])$   $∀s∈S$

````python
def generate_trajectory(env, policy, max_steps, config):
    """
    Generate a trajectory (episode) following the given policy.

    A trajectory is a sequence of (state, action, reward, next_state) tuples
    obtained by following the policy in the environment.

    Args:
        env: The environment
        policy: Policy to follow (deterministic or stochastic)
        max_steps: Maximum steps before truncating episode
        config: Configuration dictionary

    Returns:
        trajectory: List of (s, a, r, s') tuples
                   Each tuple contains:
                   - s: current state
                   - a: action taken
                   - r: reward received
                   - s': next state

    Note: Episode terminates when:
        1. Terminal state is reached (goal or hole)
        2. Maximum steps exceeded
    """
    # TODO: Implement trajectory generation
    # Hint: Use env.reset() and env.step(action)
    obs, info = env.reset(seed=config['seed'])
    s = obs['agent']
    trajectory = []
    steps = 0
    done = False
    truncated = False

    while steps < max_steps and not done and not truncated:
      a = np.argmax(policy[s])
      obs_prime, r, done, truncated, info = env.step(a)
      s_prime = obs_prime['agent']
      trajectory.append((s, a, r, s_prime))
      s = s_prime
      steps += 1

    return trajectory


def decay_alpha(initial_value, final_value, episode, max_episodes, decay_type='exponential'):
    """
    Decay the learning rate (α) over episodes.

    Two decay strategies:
    1. Exponential
    2. Linear

    Args:
        initial_value: Starting learning rate (e.g., 0.5)
        final_value: Final learning rate (e.g., 0.01)
        episode: Current episode number
        max_episodes: Total number of episodes
        decay_type: 'exponential' or 'linear'

    Returns:
        alpha: Learning rate for current episode

    Purpose: Start with large updates (fast learning) and gradually
             reduce to small updates (fine-tuning)
    """
    # TODO: Implement learning rate decay
    eps = 1e-8
    alpha = 0.0

    if decay_type == 'linear':
      alpha = initial_value + episode * (final_value - initial_value) / (max_episodes - 1 + eps)

    else:
      decay_factor = np.exp((np.log(final_value + eps) - np.log(initial_value + eps)) / (max_episodes - 1 + eps))
      alpha = initial_value * decay_factor**episode

    return alpha


def compute_true_values(env, policy, gamma, config):
    """
    Compute the true value function V^π analytically.

    This serves as ground truth to evaluate how well prediction
    methods estimate the value function.

    Args:
        env: The environment (with known dynamics env.P)
        policy: The policy to evaluate
        gamma: Discount factor
        config: Configuration dictionary

    Returns:
        V_true: True value function (numpy array)

    Note: This uses the known MDP to compute exact values.
          In practice, we wouldn't have access to this!
    """
    # TODO: Implement analytical solution
    # Hint: Use np.linalg.solve() to solve linear system
    """
    Here, I am using the below logic to find the coefficient matrix and true values:

            V[s]=∑a∈AΠ[s,a]∑(p,s',r,done)∈P[s][a]p∗(r+γ∗V[s'])  ∀s∈S

    Then Linear Equations formed are:

            0.901 x V[0] - 0.792 x V[1] - 0.099 x V[4]                     =   -0.040
            0.802 x V[1] - 0.792 x V[2]                                    =   -0.040
            0.901 x V[2] - 0.792 x V[3] - 0.099 x V[6]                     =   0.792
            0.010 x V[3]                                                   =   0.000
            -0.792 x V[0] + 0.802 x V[4]                                   =   -0.040
            1.000 x V[5]                                                   =   0.000
            -0.792 x V[2] + 0.901 x V[6] - 0.099 x V[7]                    =   -0.136
            0.010 x V[7]                                                   =   0.000
            -0.792 x V[4] + 0.901 x V[8] - 0.099 x V[9]                    =   -0.040
            -0.792 x V[8] + 0.802 x V[9]                                   =   -0.040
            -0.792 x V[6] - 0.099 x V[9] + 1.000 x V[10] - 0.099 x V[11]   =   -0.040
            -0.099 x V[7] - 0.792 x V[10] + 0.901 x V[11]                  =   -0.136
    """
    total_states = config['environment']['total_states'] + 1
    coefficient_matrix = np.zeros(shape=(total_states, total_states), dtype=np.float32)
    constant_vector = np.zeros(shape=(total_states), dtype=np.float32)

    # Constructing the coefficient matrix and contant vector
    for s in range(total_states):
      a = np.argmax(policy[s])
      coefficient_matrix[s][s] = 1
      for (p,s_prime,r,_) in env.unwrapped.P[s][a]:
        constant_vector[s] += p*r
        coefficient_matrix[s][s_prime] -= p*gamma

    if 'print_equations' in config and config['print_equations'] == True:
      # printing the linear equations formed
      width = 100
      print(f"{'Linear Equations Formed':^{width}}")
      print(f"{'-' * 80:^{width}}")

      for s in range(total_states):
        terms = []
        for s_prime in range(total_states):
          coeff = coefficient_matrix[s][s_prime]
          if coeff != 0:
            terms.append(f"{coeff:.3f} x V[{s_prime}]")

        lhs = " + ".join(terms)
        equation = f"{lhs:<65} =   {constant_vector[s]:.3f}"
        print(equation.center(width))

      print(f"{'-' * 80:^{width}}\n")

    # Solving system of linear equations
    V_true = np.linalg.solve(coefficient_matrix, constant_vector)

    return V_true
````

## Monte Carlo Prediction Methods

### Overview:

Monte Carlo methods learn from **complete episodes**. They update value estimates based on actual returns (cumulative rewards) observed.

**Key Idea:** $V(s) ≈ average\ of\ returns\ observed\ starting\ from\ s$

### Two Variants:

1. **First-Visit MC (FVMC):** Only update $V(s)$ using the return from the **first** time $s$ is visited in an episode
2. **Every-Visit MC (EVMC):** Update $V(s)$ using returns from **every** time $s$ is visited in an episode

### Update Rule:

When state s is visited (and we observe return G):
- $V(s) ← V(s) + α[G - V(s)]$

Where:
- $G = R_{t+1} + γR_{t+2} + γ²R_{t+3} + .... \ (return\ from\ time\ t)$

- $α = learning\ rate\ (decays\ over\ time)$

---


````python
def monte_carlo_prediction(env, policy, gamma, alpha_schedule, episodes, first_visit=True, config=None):
    """
    Monte Carlo Prediction (First-Visit or Every-Visit).

    This function learns the value function V^π by averaging returns
    observed from episodes generated by following policy π.

    Args:
        env: The environment
        policy: Policy to evaluate
        gamma: Discount factor (0.99)
        alpha_schedule: List of learning rates for each episode
        episodes: Number of episodes to run (500)
        first_visit: If True, use First-Visit MC; if False, use Every-Visit MC
        config: Configuration dictionary

    Returns:
        V_history: Value estimates over episodes (shape: [episodes, num_states])
                  V_history[e, s] = estimate of V(s) after episode e
        targets: MC targets (returns) for each state in each episode
    """
    # TODO: Implement MC prediction
    """
      In FVMC, for each episode we are first generating the trajectory then for each state
      within a trajectory, we are using backward pass to compute estimated expected-return
      for each time-step and then finally using forward pass updating the state-values if
      state has not been visited before.

      In EVMC, everything remains same except updating state-values for every time-step without
      checking if state has been visited before.
    """
    total_states = config['environment']['total_states'] + 1
    V_history = np.zeros(shape=(episodes, total_states), dtype=np.float32)
    V = np.zeros(shape=total_states, dtype=np.float32)
    targets = np.zeros(shape=(episodes, total_states), dtype=np.float32)
    visited = np.zeros(shape=total_states, dtype=bool)
    base_seed = config['seed']

    for e in range(episodes):
      alpha = alpha_schedule[e]
      config['seed'] = base_seed + e
      # env will get reset within generate_trajectory() with the above updated seed value.
      trajectory = generate_trajectory(env, policy, config['max_steps'], config)
      visited.fill(False)

      # computing estimated expected-return (G[t]) for each time-step using backward pass
      tlen = len(trajectory)
      G = [0] * (tlen+1)
      for t in range(tlen-1,-1,-1):
        s, a, r, s_prime = trajectory[t]
        G[t] = r + gamma * G[t+1]

      # updating the state-values (V[s]) using computed (G[t]) expected-return
      # traversing ina forward manner
      for t, (s,a,r,s_prime) in enumerate(trajectory):
        if visited[s] and first_visit:
          continue
        visited[s] = True
        V[s] = V[s] + alpha * (G[t] - V[s])
        targets[e,s] = G[t]

      V_history[e] = V
    return V_history, targets
````

## Temporal Difference (TD) Prediction

### Overview:

Temporal Difference methods learn from **individual transitions** (not complete episodes). They update value estimates based on **bootstrapping**: using current estimate of next state's value.

**Key Idea:** $V(s)$ is updated immediately after each transition

### $TD(0)$ Update Rule:

After transition $(s, a, r, s')$:
- TD Target: $target = r + γV(s')$
- TD Error: $δ = target - V(s)$
- Update: $V(s) ← V(s) + αδ$

### Advantages over MC:

1. **Online:** Can learn before episode completes
2. **Lower Variance:** Bootstrapping reduces variance (but adds bias)
3. **Works for Continuing Tasks:** Doesn't need episodes to terminate

---


````python
def td_prediction(env, policy, gamma, alpha_schedule, episodes, config=None):
    """
    Temporal Difference TD(0) Prediction.

    This function learns the value function V^π by updating estimates
    after each transition using TD error.

    Args:
        env: The environment
        policy: Policy to evaluate
        gamma: Discount factor (0.99)
        alpha_schedule: List of learning rates for each episode
        episodes: Number of episodes to run (500)
        config: Configuration dictionary

    Returns:
        V_history: Value estimates over episodes (shape: [episodes, num_states])
        targets: TD targets for each update
    """
    # TODO: Implement TD(0) prediction
    """
    In TD(0), instead of generating the trajectory, agent explores the environment
    online and then update state-values immediately without waiting for whole episode
    to end.
    """
    total_states = config['environment']['total_states'] + 1
    max_steps = config['max_steps']
    V_history = np.zeros(shape=(episodes, total_states), dtype=np.float32)
    V = np.zeros(shape=total_states, dtype=np.float32)
    targets = np.zeros(shape=(episodes, max_steps, total_states), dtype=np.float32)
    base_seed = config['seed']

    for e in range(episodes):
      alpha = alpha_schedule[e]
      # reseting the env for each episode with unique seed
      obs, info = env.reset(seed = base_seed + e)
      s = obs['agent']
      done = False
      truncated = False
      step = 0
      while not done and not truncated and step < max_steps:
        a = np.argmax(policy[s])
        # reseting the env for each episode with unique seed
        obs_prime, r, done, truncated, _ = env.step(a)
        s_prime = obs_prime['agent']
        td_target = r
        if not done:
          td_target += gamma * V[s_prime]
        td_error = td_target - V[s]
        V[s] += alpha * td_error
        targets[e,step,s] = td_target
        s = s_prime
        step += 1

      V_history[e] = V.copy()

    return V_history, targets
````

## n-Step TD Learning

### Overview:

$n-Step$ TD is a **unification** of MC and TD methods:
- $TD(0)$: Uses 1-step return (bootstraps immediately)
- MC: Uses complete return (no bootstrapping)
- n-Step TD: Uses $n-step$ return (partial bootstrapping)

### n-Step Return:

$G_t^{(n)} = R_{t+1} + γR_{t+2} + ... + γ^{n-1}R_{t+n} + γ^n V(S_{t+n})$

### Update Rule:

After observing n steps:
- $V(S_t) ← V(S_t) + α[G_t^{(n)} - V(S_t)]$

### For this problem, use n = 3:

$G_t^{(3)} = R_{t+1} + γR_{t+2} + γ²R_{t+3} + γ³V(S_{t+3})$

### Trade-off:

- Small n: Low variance, high bias (like TD)
- Large n: High variance, low bias (like MC)
- Intermediate n: Balance between bias and variance

---


````python
def n_step_td(env, policy, gamma, alpha_schedule, n, episodes, config=None):
    """
    n-Step Temporal Difference Learning.

    This function learns the value function V^π using n-step returns,
    which provides a middle ground between TD(0) and Monte Carlo.

    Args:
        env: The environment
        policy: Policy to evaluate
        gamma: Discount factor (0.99)
        alpha_schedule: List of learning rates for each episode
        n: Number of steps to look ahead (use n=3)
        episodes: Number of episodes to run (500)
        config: Configuration dictionary

    Returns:
        V_history: Value estimates over episodes (shape: [episodes, num_states])
        targets: n-step targets for each update
    """
    # TODO: Implement n-step TD with n=3
    """
    In n-Step TD, instead of updating the state-values immediately from the beginning.
    We have to first wait for n steps to start updating the state-values.
    """
    total_states = config['environment']['total_states'] + 1
    max_steps = config['max_steps']
    V_history = np.zeros(shape=(episodes, total_states), dtype=np.float32)
    V = np.zeros(shape=total_states, dtype=np.float32)
    targets = np.zeros(shape=(episodes, max_steps+n, total_states), dtype=np.float32)

    for e in range(episodes):
      alpha = alpha_schedule[e]
      # reseting the env for each episode with unique seed
      obs, _ = env.reset(seed = config['seed'] + e)
      s = obs['agent']
      step = 0
      done, truncated = False, False

      # using this 'past_states_record' list to store last n states if more than n states will be visited then we pop state from the beginning
      past_states_record = []

      while not done and not truncated and step < max_steps:
        a = np.argmax(policy[s])
        obs_prime, r, done, truncated, _ = env.step(a)
        s_prime = obs_prime['agent']

        # after taking one step, appending state_info into the 'past_states_record'
        past_states_record.append((s, r, done, s_prime))

        n_step_return = 0.0
        # checking whether agent has visited atleast n states in order to start making the updates
        if len(past_states_record) == n:

          # calculating the n_step_return for the state visited n steps before
          n_step_return = np.sum([gamma**t * r for t, (_,r,_,_) in enumerate(past_states_record)])

          # state info that was seen n steps before and popping its info from 'past_states_record'
          (rs, rr, _, rs_prime) = past_states_record.pop(0)

          if not done:
            n_step_return += gamma**n * V[s_prime]
          V[rs] += alpha * (n_step_return - V[rs])
          targets[e,step,rs] = n_step_return

        s = s_prime
        step += 1

      # after the episode ends, we are processing the leftover states within 'past_states_record' in order to update their state-values as well.
      while len(past_states_record) > 0:
        n_step_return = np.sum([gamma**t * r for t, (_,r,_,_) in enumerate(past_states_record)])
        (rs, rr, _, rs_prime) = past_states_record.pop(0)
        V[rs] += alpha * (n_step_return - V[rs])
        targets[e,step,rs] = n_step_return
        step += 1

      V_history[e] = V.copy()

    return V_history, targets
````

## $TD(λ)$ with Eligibility Traces

### Overview:

$TD(λ)$  is an **elegant unification** of all $n-step$ methods using **eligibility traces**.

**Key Innovation:** Instead of choosing one value of $n,\ TD(λ)$ combines returns from **all n**-step methods using parameter $λ ∈ [0, 1]$.

### Eligibility Traces:

Each state maintains an **eligibility trace** $e(s)$ that:
- **Increases** when state is visited
- **Decays** over time by factor $γλ$

Update rules:
- **Trace Update:** $e(s) ← γλ~e(s) + 1~$  (when s is visited)
- **Value Update:** $V(s) ← V(s) + αδ~e(s)~$
  (for all s)

Where $δ = r + γV(s') - V(s)~$ is the TD error.

### $λ$ Parameter:

- $λ = 0$: Reduces to $TD(0)~$ (only current state updated)
- $λ = 1$: Similar to MC (all visited states updated equally)
- $λ = 0.3$: **Use this value** (balances bias-variance trade-off)

### Why Eligibility Traces?

1. **Credit Assignment:** States visited long ago still get updates (decaying)
2. **Computational Efficiency:** Update all states in $O(n)$ time, not $O(n²)$
3. **Online Learning:** Works with continuing tasks

---


````python
def td_lambda(env, policy, gamma, alpha_schedule, lambda_val, episodes, config=None):
    """
    TD(λ) Prediction with Eligibility Traces (Backward View).

    This function learns the value function V^π using eligibility traces,
    which provides an efficient way to combine n-step returns for all n.

    Args:
        env: The environment
        policy: Policy to evaluate
        gamma: Discount factor (0.99)
        alpha_schedule: List of learning rates for each episode
        lambda_val: Trace decay parameter (use λ=0.3)
        episodes: Number of episodes to run (500)
        config: Configuration dictionary

    Returns:
        V_history: Value estimates over episodes (shape: [episodes, num_states])
        eligibility_traces: Traces for specific episodes (for visualization)
        targets: TD targets for updates

    Important: Store eligibility traces for episode 100 for plotting!
    """
    # TODO: Implement TD(λ) with λ=0.3
    # Remember to save eligibility traces for episode 100!

    """
    TD(λ) extends the idea of TD(0) by introducing ELigibility traces whose main purpose is to
    give higher credit to more frequently visited states and lower to least frequently visited states.
    """
    total_states = config['environment']['total_states'] + 1
    max_steps = config['max_steps']
    V_history = np.zeros(shape=(episodes, total_states), dtype=np.float32)
    V = np.zeros(shape=total_states, dtype=np.float32)
    eligibility_traces = np.zeros(shape=(max_steps,total_states), dtype=np.float32)
    targets = np.zeros(shape=(episodes, max_steps, total_states), dtype=np.float32)
    E = np.zeros(shape=total_states, dtype=np.float32)

    et_episode_to_track = 100

    for e in range(episodes):
      alpha = alpha_schedule[e]
      # reseting the env for each episode with unique seed
      obs, info = env.reset(seed = config['seed'] + e)
      s = obs['agent']
      done = False
      truncated = False
      step = 0
      E.fill(0)
      while not done and not truncated and step < max_steps:
        # agent is taking one step into the env
        a = np.argmax(policy[s])
        obs_prime, r, done, truncated, _ = env.step(a)
        s_prime = obs_prime['agent']
        td_target = r
        if not done:
          td_target += gamma * V[s_prime]
        td_error = td_target - V[s]

        # before updating the state-values, first giving credit to the state 's' and then
        # updaing the state-values for all the states based on the credit assignment till now.
        # and finally decaying credits with a factor of γλ in each step.
        E[s] += 1
        V = V + alpha * td_error * E
        E = gamma * lambda_val * E

        targets[e,step,s] = td_target
        if e == et_episode_to_track:
          eligibility_traces[step] = E.copy()
        s = s_prime
        step += 1

      V_history[e] = V.copy()

    return V_history, eligibility_traces, targets
````

### Plot 12: Comparison of All Methods (5 points)

Create a single plot comparing the **convergence speed** of all five methods.

**Suggested Metric:** Root Mean Square Error (RMSE) from true values

$RMSE_{(episode)} = sqrt\ ( mean\ over\ states\ [(V_{estimated(s)} - V_{true(s)})²] )$

**Plot Specifications:**
- X-axis: Episode number
- Y-axis: RMSE from true values
- One line for each method (5 lines total)
- Log scale on both axes recommended
- Clear legend distinguishing methods

**What to Compare:**
- Which method converges fastest?
- Which method has lowest final error?
- Which method is most stable (least variance)?


````python
# TODO: Generate Plot 12 - Comparison
# Calculate RMSE for each method at each episode
# Plot all methods on same axes
def plot_convergence_speed(stacked_V_history, V_true, config_list):
  num_of_algo, total_episodes, total_states = stacked_V_history.shape
  rmse = np.zeros(shape=(num_of_algo, total_episodes), dtype=np.float32)
  non_terminal_states = list(filter(lambda state: state not in [3, 5, 7], range(total_states)))

  for a in range(num_of_algo):
    for e in range(total_episodes):
      rmse[a,e] = np.sqrt(np.mean(np.square(stacked_V_history[a,e,non_terminal_states] - V_true[non_terminal_states])))

  plt.rcParams.update({
        "font.size": 12,
        "axes.labelsize": 13,
        "axes.titlesize": 14,
        "legend.fontsize": 9,
        "xtick.labelsize": 10,
        "ytick.labelsize": 10,
    })

  plt.figure(figsize=(10,6))
  for a in range(num_of_algo):
    plt.plot(rmse[a], label=config_list[a]['method'])

  plt.title(f"[FVMC, EVMC, TD(0), N-StepTD, TD(λ)]\nRMSE over {total_episodes} episodes")
  plt.xlabel("Episode (log scale)")
  plt.ylabel("RMSE from true values (log scale)")
  plt.xscale('log')
  plt.yscale('log')
  plt.legend()
  plt.grid(True, linestyle="dotted", alpha=0.4)
  plt.tight_layout()

  plt.savefig(f"convergence_comparison.pdf", bbox_inches="tight")
  plt.show()

stacked_V_history = np.stack([fvmc_V_history, evmc_V_history, td_V_history, ntd_V_history, td_lambda_V_history])
config_list = [fvmc_config, evmc_config, td_config, ntd_config, td_lambda_config]
plot_convergence_speed(stacked_V_history, V_true, config_list)
````

2. All the methods (FVMC, EVMC, TD(0), n-Step TD and TD(λ)) have the lowest final error.
3. TD(λ) and TD(0) are most stable.

### Comprehensive analysis:

1. **Convergence Speed:**
  > FVMC, EVMC and n-Step TD convereges the fastest because they are using either full or partial expected return for updating the state-value, making them less biased and allowing them to converge the fastest.
  >
  > TD(0) is the slowest because it uses bootstrapping to update its state-values at each step, in order to reduce variance but for the sake of lower convergence speed.

2. **Bias-Variance Trade-off:**
  > It is evident from the Value Estimation plots that even after convergence,  MC methods become stable much later as it shows noticeable spikes throughout all the episodes but less biased because of using full expected return.
  >
  > On the other hand, it is observed that TD methods have more bias (farther from the true values) but less variance (less spikes) because of bootstrapping.

3. **Practical Recommendations:**
  > Use MC when the observation space is small, as it relies on full episode returns and storing visited states; it is suitable for complex or stochastic environments and handles noisy/outlier returns robustly.
  >
  > Use TD when the environment is simpler, as it updates via bootstrapped estimates and can be sensitive to outliers.
  >
  > n-Step TD balances between MC and TD(0), performing well in both settings, but its effectiveness depends on choosing an appropriate n.
  >
  > TD(λ) provides a better trade-off between MC and TD(0), reducing bias and variance, making it robust to outliers and stable even in large observation spaces for both stochastic and deterministic environments.

  > ***Computational trade off:***
  >> MC > n-Step TD ≡ TD(λ) > TD(0)

4. **Hyperparameter Sensitivity:**
  >
  ****Choice of α decay:****
  > α ∈ $[0,1]$: As the value of α increases, learning becomes faster but introduces more spikes in the estimates. For non-convex systems, it might jump out of global optimum region, resulting in sub-optimal estimates.
  >
  ****Choice of n and λ:****
  >
  > Both n and λ are hyperparameters that manage the trade-off between MC and TD(0).
  >> n ∈ $[1,∞]$ while λ ∈ $[0,1]$
  >>
  >> As n (or λ) increases, the update behavior gradually transitions from TD(0)-like updates toward MC-like updates.
